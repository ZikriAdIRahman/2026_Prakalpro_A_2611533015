## 1.1 Latar Belakang

Dalam pemrograman, percabangan (kondisional) merupakan struktur kontrol yang digunakan untuk mengambil keputusan berdasarkan suatu kondisi tertentu. Kondisi tersebut dievaluasi menjadi nilai boolean (True/False), dan alur program akan mengikuti blok pernyataan yang sesuai dengan hasil evaluasi. Percabangan menjadi elemen dasar yang selalu digunakan dalam hampir setiap program, mulai dari validasi input pengguna hingga penentuan logika bisnis yang kompleks.

Python menyediakan berbagai bentuk percabangan, di antaranya if tunggal untuk mengeksekusi satu blok pernyataan ketika kondisi benar, if-else untuk dua kemungkinan alur, if-elif-else untuk lebih dari dua kemungkinan kondisi, multi-if terpisah (bukan elif) agar beberapa kondisi dapat dipenuhi sekaligus (seperti akumulasi diskon), serta match-case untuk mencocokkan nilai dengan beberapa pola secara lebih rapi. Oleh karena itu, praktikum tentang Percabangan ini dilakukan untuk memberikan pemahaman mendalam mengenai berbagai jenis struktur percabangan dalam Python serta cara penggunaannya dalam program sederhana.

## 2.1 Dasar Teori

Praktikum Percabangan membahas berbagai jenis struktur kondisi dalam Python serta cara penggunaannya untuk mengendalikan alur program.

### A. If Tunggal
Digunakan untuk mengeksekusi satu blok pernyataan hanya jika kondisi bernilai True. Jika kondisi bernilai False, blok tersebut dilewati dan program melanjutkan ke pernyataan berikutnya.

### B. If-Else
Digunakan untuk dua kemungkinan alur program: blok `if` dijalankan jika kondisi bernilai True, sedangkan blok `else` dijalankan jika kondisi bernilai False.

### C. If-Elif-Else dan Operator Logika
Digunakan untuk menangani lebih dari dua kemungkinan kondisi. Kondisi majemuk dibentuk menggunakan operator logika `and` (kedua kondisi harus benar), `or` (salah satu kondisi benar), serta `!=` (ketidaksamaan) untuk mengevaluasi beberapa syarat dalam satu percabangan.

### D. Multi-If Terpisah
Serangkaian blok `if` yang independen (bukan elif) sehingga setiap kondisi diperiksa secara terpisah dan beberapa kondisi dapat terpenuhi sekaligus. Teknik ini digunakan untuk mengakumulasikan diskon yang dapat ditumpuk.

### E. Match-Case
Digunakan untuk mencocokkan nilai variabel dengan sejumlah pola (case). Setiap case berisi pernyataan yang dijalankan jika nilainya cocok, dan `case _` (default) dijalankan apabila tidak ada pola yang cocok.

Seluruh alur program pada praktikum ini disusun murni menggunakan pilar-pilar percabangan tersebut tanpa melibatkan nested-if (if di dalam if).

## 2.2 Kode Program

### 2.2.1 Program if1_2611533015 (If Tunggal)

```python
#Buat nama program untuk kondisional if 
#Program ini menggunakan fungsi input()

ipk_3015 = float(input("Input IPK Anda = "))
if ipk_3015 > 2.75:
    print("Anda lulus sangat memuaskan dengan IPK" + str(ipk_3015))

print("Program Selesai")
```

**Penjelasan Program if1_2611533015:**

Program ini mendemonstrasikan **if tunggal** dalam Python, yaitu struktur percabangan satu arah yang hanya menjalankan blok pernyataan ketika kondisi bernilai True.

1. **`ipk_3015 = float(input("Input IPK Anda = "))`** — Menerima masukan IPK dari pengguna (berupa string) lalu dikonversi menjadi tipe data float menggunakan `float()` agar nilai pecahan dapat dibandingkan secara numerik.
2. **`if ipk_3015 > 2.75:`** — Kondisi percabangan if tunggal. Ekspresi `ipk_3015 > 2.75` bernilai True apabila IPK pengguna lebih besar dari 2.75.
3. **`print("Anda lulus sangat memuaskan dengan IPK" + str(ipk_3015))`** — Blok pernyataan yang hanya dijalankan ketika kondisi di atas bernilai True. Operator `+` menggabungkan string dengan hasil konversi `str(ipk_3015)`.
4. **`print("Program Selesai")`** — Berada di luar blok `if` (tidak menjorok), sehingga selalu dijalankan terlepas dari hasil kondisi.

Program ini menunjukkan bahwa if tunggal tidak memiliki cabang alternatif; apabila kondisi bernilai False, program langsung melanjutkan ke pernyataan berikutnya.

### 2.2.2 Program if2_2611533015 (If-Else)

```python
# Buat file dengan nama if2_2611533015.py
# Buat program untuk kondisional If
# Nama variabel ditambah 4 digit nim terakhir contoh ipk_1234
# Program ini menggunakan fungsi input()    

ipk_3015 = float(input("Input IPK Anda: "))

if ipk_3015 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK:", str(ipk_3015))

else:
    print("Anda Tidak Lulus")
print("Program Selesai")
```

**Penjelasan Program if2_2611533015:**

Program ini mendemonstrasikan **if-else** dalam Python, yaitu struktur percabangan dua arah yang menyediakan cabang alternatif ketika kondisi utama bernilai False.

1. **`ipk_3015 = float(input("Input IPK Anda: "))`** — Menerima masukan IPK dan mengonversinya menjadi tipe data float.
2. **`if ipk_3015 > 2.75:`** — Kondisi utama. Apabila IPK lebih besar dari 2.75, blok setelah `if` yang dijalankan.
3. **`print("Anda Lulus Sangat Memuaskan dengan IPK:", str(ipk_3015))`** — Blok True, dicetak ketika IPK > 2.75.
4. **`else:`** — Cabang alternatif. Blok ini dijalankan hanya ketika kondisi utama bernilai False (IPK <= 2.75), sehingga mencetak "Anda Tidak Lulus".
5. **`print("Program Selesai")`** — Berada di luar blok percabangan, sehingga selalu dijalankan.

Perbedaan utama dengan if tunggal: pada if-else, salah satu dari dua blok pasti dijalankan sehingga tidak ada kemungkinan program melewati percabangan tanpa mengeksekusi pernyataan apa pun.

### 2.2.3 Program if_elif_else1_2611533015 (If-Elif-Else + Operator Logika)

```python
# Buat file dengan nama if_elif_else1_2611533015.py
# Buat program untuk kondisional If
# Nama variabel ditambah 4 digit nim terakhir contoh ipk_1234
# Program ini menggunakan fungsi input()

umur_3015 = int(input("Input umur Anda: "))
sim_3015 = input("Apakah Anda Sudah Punya SIM C (y/t): ")[0]

if umur_3015 >= 17 and sim_3015 == "y": 
    print("Anda sudah dewasa dan boleh bawa motor")

elif umur_3015 >= 17 and sim_3015 != "y":
    print("Anda sudah dewasa tetapi tidak boleh bawa motor")

elif umur_3015 < 17 and sim_3015 == "y":
    print("Anda belum cukup umur punya SIM")

else:
    print("Anda belum cukup umur dan tidak boleh bawa motor")
print("Program Selesai")
```

**Penjelasan Program if_elif_else1_2611533015:**

Program ini mendemonstrasikan **if-elif-else** dengan **kondisi majemuk (operator logika)** pada pemeriksaan kelayakan SIM.

1. **`umur_3015 = int(input("Input umur Anda: "))`** — Menerima umur pengguna dan mengonversinya menjadi integer.
2. **`sim_3015 = input("...")[0]`** — Menerima jawaban SIM C, lalu `[0]` mengambil karakter pertama, sehingga input "ya" tetap terbaca sebagai "y".
3. **`if umur_3015 >= 17 and sim_3015 == "y":`** — Operator `and` mengharuskan kedua kondisi benar sekaligus: sudah dewasa (umur >= 17) DAN memiliki SIM (sim == "y").
4. **`elif umur_3015 >= 17 and sim_3015 != "y":`** — Operator `!=` memeriksa bahwa pengguna sudah dewasa tetapi tidak memiliki SIM (sim bukan "y"), sehingga tidak boleh membawa motor.
5. **`elif umur_3015 < 17 and sim_3015 == "y":`** — Kondisi umur belum 17 tetapi mengklaim punya SIM; karena belum cukup umur, identitas dianggap tidak valid.
6. **`else:`** — Cabang terakhir untuk pengguna yang belum cukup umur dan tidak memiliki SIM.
7. Rantai **if-elif-else** berhenti pada blok pertama yang kondisinya bernilai True, sehingga tidak ada kemungkinan mencetak dua status sekaligus.

### 2.2.4 Program nilai_2611533015 (If-Elif-Else Konversi Nilai)

```python
# Buat file dengan nama nilai_nim.py
# Buat program untuk kondisional If
# Nama variabel ditambah 4 digit nim terakhir contoh ipk_1234
# Program ini menggunakan fungsi input()
# Program konversi nilai angka ke huruf

nilai_3015 = int(input("Masukkan Nilai Angka: "))

if nilai_3015 >= 81:
    print("A")
elif nilai_3015 >= 70:
    print("B")
elif nilai_3015 >= 60:
    print("C")
elif nilai_3015 >= 50:
    print ("D")
else:
    print("E")
```

**Penjelasan Program nilai_2611533015:**

Program ini mendemonstrasikan **if-elif-else** bertingkat untuk mengonversi nilai angka menjadi nilai huruf.

1. **`nilai_3015 = int(input("Masukkan Nilai Angka: "))`** — Menerima nilai angka dan mengonversinya menjadi integer.
2. **`if nilai_3015 >= 81:`** — Jika nilai >= 81, mencetak "A", lalu rantai percabangan berhenti dan blok elif/else berikutnya tidak diperiksa.
3. **`elif nilai_3015 >= 70:`, `elif nilai_3015 >= 60:`, `elif nilai_3015 >= 50:`** — Blok alternatif yang hanya diperiksa apabila semua kondisi sebelumnya bernilai False. Karena urutannya menurun dari besar ke kecil, nilai 75 misalnya jatuh tepat pada blok "B".
4. **`else:`** — Dijalankan untuk nilai kurang dari 50, sehingga mencetak "E".

Program ini menunjukkan bahwa pada rantai if-elif-else hanya blok pertama yang kondisinya True yang dieksekusi, sehingga satu rentang nilai tidak mungkin tercetak dua kali.

### 2.2.5 Program match_case_2611533015 (Match-Case)

```python
# Buat File dengan nama match_case.py
# Buat program untuk match case
# Nama variabel ditambah 4 digit nim terakhir contoh ipk_1234
# Program ini menggunakan fungsi input()
# Program konversi angka menjadi nama bulan

bulan_3015 = int(input("Masukkn Angka bulan (1-12: )"))

match bulan_3015:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")

print("Program selesai")
```

**Penjelasan Program match_case_2611533015:**

Program ini mendemonstrasikan **match-case** dalam Python, yaitu struktur percabangan yang mencocokkan nilai variabel dengan sejumlah pola (case).

1. **`bulan_3015 = int(input(...))`** — Menerima angka bulan dan mengonversinya menjadi integer.
2. **`match bulan_3015:`** — Nilai variabel `bulan_3015` dibandingkan dengan setiap pola `case` secara berurutan dari atas.
3. **`case 1:`** s.d. **`case 12:`** — Setiap case mencetak nama bulan yang sesuai dengan angkanya. Hanya case yang nilainya cocok yang dijalankan.
4. **`case _:`** — Pola default (catch-all) yang dijalankan apabila tidak ada case yang cocok (misalnya angka 13 atau 0), sehingga mencetak "Angka tidak valid".
5. **`print("Program selesai")`** — Berada di luar blok match-case, sehingga selalu dijalankan.

Program ini menunjukkan bahwa match-case membuat pencocokan nilai menjadi lebih rapi dan mudah dibaca dibandingkan rantai if-elif-else yang panjang, terutama ketika harus menangani banyak kemungkinan nilai.

### 2.2.6 Program multi_if1_2611533015 (Multi-If Terpisah)

```python
# Buat File dengan nama multi_if.py
# Buat program untuk kondisional If
# Nama variabel ditambah 4 digit nim terakhir contoh ipk_1234
# Program ini menggunakan fungsi input()

umur_3015 = int(input("Input umur Anda: "))
sim_3015 = input("Apakah Anda Sudah Punya SIM C (y/t): ")[0]

if umur_3015 >= 17 and sim_3015 == "y": 
    print("Anda sudah dewasa dan boleh bawa motor")

if umur_3015 >= 17 and sim_3015 != "y":
    print("Anda sudah dewasa tetapi tidak boleh bawa motor")

if umur_3015 < 17 and sim_3015 != "y":
    print("Anda belum cukup umur bawa motor")

if umur_3015 < 17 and sim_3015 == "y":
    print("Anda belum cukup umur punya SIM")
```

**Penjelasan Program multi_if1_2611533015:**

Program ini mendemonstrasikan **multi-if terpisah** dalam Python, yaitu beberapa blok `if` independen yang masing-masing diperiksa sendiri-sendiri (bukan elif).

1. **`umur_3015 = int(input("Input umur Anda: "))`** — Menerima umur dan mengonversinya menjadi integer.
2. **`sim_3015 = input("...")[0]`** — Mengambil karakter pertama dari jawaban SIM C.
3. **Empat blok `if` terpisah** — Setiap blok memiliki kondisi lengkap yang dibentuk dengan operator `and` dan `!=`. Kondisi saling meniadakan sehingga untuk satu pasang input hanya satu blok yang bernilai True, namun semua blok tetap diperiksa satu per satu.
4. Berbeda dengan if-elif-else, pada multi-if **setiap kondisi selalu dievaluasi** tanpa terhenti oleh kondisi sebelumnya. Inilah dasar teknik akumulasi kondisi seperti pada program berikutnya.

Program ini menunjukkan bahwa multi-if berguna ketika beberapa kondisi dapat bernilai True secara bersamaan dan seluruh blok yang memenuhi tetap harus diproses.

### 2.2.7 Program multi_if2_2611533015 (Multi-If Akumulasi Diskon)

```python
# Buat file dengan nama multi_if2_2611533015.py
# Buat program untuk kondisional If
# Nama variabel ditambah 4 digit nim terakhir contoh ipk_1234
# Program ini menggunakan fungsi input()
# Program Menghitung Diskon belanja

# Input dari user
total_belanja_3015 = float(input("Input Total Belanja (Rp): "))

# Input status member
input_member_3015 = input("Apakah Anda Member (y/t): ").strip().lower()
is_member_3015 = input_member_3015 in ["y", "ya"]

#Input status kode promo
input_promo_3015 = input("Apakah Kode Promo valid (y/t): ").strip().lower()
kode_promo_valid_3015 = input_promo_3015 in ["y", "ya"]

total_diskon_persen_3015 = 0

# MULTI-IF terpisah Setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk jika memenuhi beberapa syarat sekaligus

if total_belanja_3015 >= 1000000:
    total_diskon_persen_3015 += 10  # Diskon 10% untuk belanja >= 1 juta

if is_member_3015:
    total_diskon_persen_3015 += 5  # Diskon tambahan 5% untuk member

if kode_promo_valid_3015:
    total_diskon_persen_3015 += 15  # Diskon tambahan 15% untuk kode promo valid

# Menghitung nominal diskon dan total bayar
nominal_diskon_3015 = (total_diskon_persen_3015 / 100) * total_belanja_3015
total_bayar_3015 = total_belanja_3015 - nominal_diskon_3015

# Output hasil perhitungan diskon
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon : {total_diskon_persen_3015}% (Rp {nominal_diskon_3015:.0f})")
print(f"Total Bayar   : Rp {total_bayar_3015:.0f}")

print(f"Total Diskon yang anda dapatkan : {total_diskon_persen_3015}%")
#Output: Total diskon yang anda dapatkan : 30% jika belanja > 1 juta, member, dan kode promo valid
```

**Penjelasan Program multi_if2_2611533015:**

Program ini mendemonstrasikan **multi-if terpisah untuk akumulasi diskon bertingkat**. Karena setiap blok menggunakan `if` (bukan elif), seluruh kondisi yang terpenuhi akan dijumlahkan (ditumpuk) pada variabel `total_diskon_persen_3015`.

1. **`total_belanja_3015 = float(input(...))`** — Menerima total belanja dan mengonversinya menjadi float.
2. **`.strip().lower()`** — Input status member dan kode promo dibersihkan dari spasi (strip) dan diubah menjadi huruf kecil (lower), lalu diterjemahkan menjadi boolean melalui `... in ["y", "ya"]`.
3. **`if total_belanja_3015 >= 1000000:`** — Diskon 10% ditambahkan apabila belanja setidaknya 1 juta (Diskon Belanja Besar).
4. **`if is_member_3015:`** — Diskon 5% ditambahkan apabila pengguna berstatus member (Diskon Member).
5. **`if kode_promo_valid_3015:`** — Diskon 15% ditambahkan apabila kode promo valid (Diskon Voucher Promo).
6. **Karena ketiganya `if` independen**, pengguna yang belanjanya >= 1 juta, berstatus member, dan memiliki kode promo valid memperoleh diskon total 10% + 5% + 15% = 30%.
7. **`nominal_diskon_3015`** dan **`total_bayar_3015`** dihitung berdasarkan total persen diskon yang terkumpul, lalu ditampilkan menggunakan f-string.

Program ini menunjukkan bahwa multi-if terpisah sangat tepat untuk diskon yang dapat diakumulasikan, karena memungkinkan beberapa bonus berjalan sekaligus tanpa saling mengesampingkan.

Kelayakan pengendara diverifikasi menggunakan if-elif-else dan operator logika. Diskon diakumulasikan dengan multi-if terpisah (Diskon Belanja Besar +10%, Diskon Member +5%, Diskon Voucher Promo +15%, Diskon Tambahan Rombongan +5%), kemudian kelulusan audit (bonus souvenir apabila total bayar lebih dari Rp300.000) dievaluasi menggunakan if-else.

## Kode Jawaban :

### a. Kode Program sistem_loket_2611533015

```python
# Sistem Loket Terpadu & Audit Transaksi Ekspedisi Wahana
# Disusun murni menggunakan pilar percabangan Pekan 4:
#   1. if tunggal           -> validasi kelogisan kuota tiket
#   2. if-elif-else         -> validasi izin kendali wahana (operator logika and, !=)
#   3. match-case (1-5, _)  -> pemilihan paket wahana
#   4. multi-if terpisah    -> akumulasi diskon bertingkat (diskon ditumpuk)
#   5. if-else              -> evaluasi kelulusan audit (bonus souvenir)
# Tidak ada nested-if (if di dalam if).
# Nama variabel WAJIB diakhiri 4 digit NIM terakhir (_3015), format snake_case.

print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")

# ============ 1. INPUT DATA PENGUNJUNG & STRING HANDLING ============
nama_3015 = input("Masukkan Nama Pengunjung        : ")
umur_3015 = int(input("Input umur anda                 : "))
sim_3015 = input("Apakah Anda Sudah Punya SIM C (y/t): ").strip().lower()[0]

# Menu paket wahana
print("\nPilihan Paket Wahana (1-5):")
print("  1. Safari Rimba         (Rp 50,000)")
print("  2. Arung Jeram          (Rp 75,000)")
print("  3. Motor ATV Ekstrim    (Rp 120,000)")
print("  4. Roller Coaster Kilat (Rp 100,000)")
print("  5. All-Access VIP       (Rp 220,000)")

paket_3015 = int(input("Masukkan nomor paket (1-5)      : "))
jumlah_tiket_3015 = int(input("Masukkan jumlah tiket           : "))

# ---- IF TUNGGAL: validasi kelogisan kuota tiket ----
if jumlah_tiket_3015 <= 0:
    print("Peringatan: Kuota tiket tidak valid!")

is_member_3015 = input("Apakah Anda member? (y/t)       : ").strip().lower()
kode_promo_valid_3015 = input("Apakah kode promo valid? (y/t)  : ").strip().lower()

# ============ 2. PEMILIHAN WAHANA MENGGUNAKAN MATCH-CASE (1-5 DAN _) ============
match paket_3015:
    case 1:
        nama_paket_3015 = "Wahana Safari Rimba"
        harga_satuan_3015 = 50000
    case 2:
        nama_paket_3015 = "Wahana Arung Jeram"
        harga_satuan_3015 = 75000
    case 3:
        nama_paket_3015 = "Wahana Motor ATV Ekstrim"
        harga_satuan_3015 = 120000
    case 4:
        nama_paket_3015 = "Wahana Roller Coaster Kilat"
        harga_satuan_3015 = 100000
    case 5:
        nama_paket_3015 = "Wahana All-Access VIP"
        harga_satuan_3015 = 220000
    case _:
        print("Paket wahana tidak valid!")
        exit()

# ============ 3. VALIDASI IZIN KENDALI WAHANA (IF-ELIF-ELSE + and / !=) ============
print("\n--- KELAYAKAN PENGENDARA WAHANA ---")
# Rantai if-elif-else tunggal tanpa nested-if.
# Untuk paket 3 (Motor ATV Ekstrim) evaluasi izin pengendara,
# sedangkan untuk paket selain 3 cukup cek umur >= 10 tahun.
if paket_3015 == 3 and umur_3015 >= 17 and sim_3015 == "y":
    print("Status Akses: Anda sudah dewasa dan boleh mengendarai ATV sendiri.")
elif paket_3015 == 3 and umur_3015 >= 17 and sim_3015 != "y":
    print("Status Akses: Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur).")
elif paket_3015 == 3 and umur_3015 < 17 and sim_3015 == "y":
    print("Status Akses: Identitas tidak valid: Belum cukup umur memiliki SIM.")
elif paket_3015 == 3:
    print("Status Akses: Anda belum cukup umur dan tidak boleh bawa motor ATV.")
elif umur_3015 >= 10:
    print(f"Status Akses: Umur memenuhi syarat (>= 10 tahun). {nama_3015} dipersilakan menikmati {nama_paket_3015}.")
else:
    print("Status Akses: Maaf, umur belum mencukupi (minimal 10 tahun) untuk wahana ini.")

# ============ 4. AKUMULASI DISKON BERTINGKAT (MULTI-IF TERPISAH) ============
subtotal_3015 = harga_satuan_3015 * jumlah_tiket_3015
total_diskon_persen_3015 = 0

# Setiap blok if independen (bukan elif) sehingga diskon dapat diakumulasikan/ditumpuk.
if subtotal_3015 >= 200000:
    total_diskon_persen_3015 += 10  # Diskon Belanja Besar

if is_member_3015 in ["y", "ya"]:
    total_diskon_persen_3015 += 5  # Diskon Member

if kode_promo_valid_3015 in ["y", "ya"]:
    total_diskon_persen_3015 += 15  # Diskon Voucher Promo

if jumlah_tiket_3015 >= 5:
    total_diskon_persen_3015 += 5  # Diskon Tambahan Rombongan

# ============ 5. EVALUASI KELULUSAN AUDIT (IF-ELSE) ============
nominal_diskon_3015 = subtotal_3015 * (total_diskon_persen_3015 / 100)
total_bayar_3015 = subtotal_3015 - nominal_diskon_3015

print("\n--- Rincian Pembayaran ---")
print(f"Subtotal Belanja : Rp {subtotal_3015:,.0f}")
print(f"Total Diskon     : {total_diskon_persen_3015}% (Rp {nominal_diskon_3015:,.0f})")
print(f"Total Bayar      : Rp {total_bayar_3015:,.0f}")

if total_bayar_3015 > 300000:
    print("Catatan Layanan  : Selamat! Anda berhak mendapatkan Souvenir Gratis.")
else:
    print("Catatan Layanan  : Terima kasih telah berkunjung.")

print("Program Selesai")
```

### b. Jawaban Pertanyaan Analisis

**1. Bagaimana if tunggal diterapkan pada program dan apa outputnya ketika jumlah tiket tidak valid?**

If tunggal diterapkan tepat setelah input jumlah tiket, yaitu `if jumlah_tiket_3015 <= 0:` yang mencetak peringatan "Peringatan: Kuota tiket tidak valid!" apabila pengunjung memasukkan jumlah tiket 0 atau bernilai negatif. If tunggal ini tidak memiliki cabang alternatif; sisa program tetap berjalan setelah peringatan dicetak.

**2. Jelaskan peran match-case dalam menentukan jenis wahana dan apa yang terjadi pada case _.**

`match paket_3015` mencocokkan nomor paket yang dimasukkan pengunjung dengan pola `case 1` sampai `case 5`. Setiap case hanya menetapkan `nama_paket_3015` beserta `harga_satuan_3015` paket yang bersangkutan. Apabila nomor paket berada di luar rentang 1-5, pola `case _` (default) dijalankan, mencetak "Paket wahana tidak valid!" lalu menghentikan eksekusi transaksi dengan `exit()`, sehingga rincian pembayaran tidak dihitung.

**3. Bagaimana operator logika and dan != digunakan pada validasi izin kendali paket Motor ATV Ekstrim?**

Kondisi `paket_3015 == 3 and umur_3015 >= 17 and sim_3015 == "y"` menggunakan operator `and` yang menuntut seluruh syarat benar agar pengunjung "sudah dewasa dan boleh mengendarai ATV sendiri". Sementara kondisi `... and sim_3015 != "y"` memanfaatkan operator `!=` (tidak sama) untuk menandai pengunjung dewasa yang belum memiliki SIM sehingga "wajib didampingi instruktur". Rantai if-elif-else ini memastikan hanya satu status akses yang tercetak untuk satu input.

**4. Mengapa multi-if diperlukan untuk menghitung diskon akumulatif dan mengapa tidak menggunakan elif?**

Multi-if diperlukan karena setiap blok `if` diperiksa secara independen, sehingga seluruh kondisi yang terpenuhi diakumulasikan pada `total_diskon_persen_3015`. Apabila digunakan elif, hanya blok pertama yang kondisinya True yang dihitung sehingga diskon tidak dapat ditumpuk. Pada contoh program, subtotal Rp240.000 mengaktifkan Diskon Belanja Besar (+10%), pengunjung member (+5%), dan kode promo valid (+15%), sehingga total diskon menjadi 30% sekaligus.

**5. Jelaskan alur evaluasi kelulusan audit (if-else) pada program.**

Setelah `nominal_diskon_3015 = subtotal_3015 * (total_diskon_persen_3015 / 100)` dan `total_bayar_3015 = subtotal_3015 - nominal_diskon_3015` dihitung, struktur if-else mengevaluasi kondisi `total_bayar_3015 > 300000`. Apabila lebih dari Rp300.000, dicetak bonus "Selamat! Anda berhak mendapatkan Souvenir Gratis.", sedangkan pada kondisi lainnya (else) dicetak "Terima kasih telah berkunjung." — salah satu dari dua pesan inilah yang menjadi catatan layanan pada rincian pembayaran.

**6. Identifikasi minimal satu kondisi yang menghasilkan True dan satu yang menghasilkan False pada contoh output program.**

Pada contoh output (subtotal Rp240.000): kondisi `subtotal_3015 >= 200000` menghasilkan **True** (240.000 >= 200.000) sehingga memberikan diskon +10%, sedangkan kondisi `jumlah_tiket_3015 >= 5` menghasilkan **False** (jumlah tiket hanya 2) sehingga Diskon Tambahan Rombongan tidak diaktifkan. Hasil True/False ini menentukan besar persen diskon yang terakumulasi pada `total_diskon_persen_3015`.
