## 2.1 Dasar Teori

Praktikum Operator membahas berbagai jenis operator dalam Python serta cara penggunaannya untuk mengolah data.

### A. Operator Aritmatika
Digunakan untuk melakukan perhitungan matematis, meliputi penjumlahan (`+`), pengurangan (`-`), perkalian (`*`), pembagian (`/`), pembagian bulat (`//`), sisa bagi/modulus (`%`), dan pangkat (`**`).

### B. Operator Assignment
Digunakan untuk memberikan nilai ke variabel, meliputi assignment biasa (`=`) dan assignment gabungan seperti `+=`, `-=`, `*=`, `/=`, `//=`, `%=`, dan `**=` yang merupakan bentuk singkat dari operasi diikuti assignment.

### C. Operator Perbandingan
Digunakan untuk membandingkan dua nilai dan menghasilkan nilai boolean (True/False), meliputi `>`, `<`, `>=`, `<=`, `==`, dan `!=`. Python juga mendukung perbandingan berantai seperti `0 < x < 100`.

### D. Operator Logika
Digunakan untuk menggabungkan beberapa ekspresi boolean, meliputi `and` (konjungsi), `or` (disjungsi), dan `not` (negasi).

### E. Operator Bitwise
Digunakan untuk memanipulasi data pada level bit (biner), meliputi AND (`&`), OR (`|`), XOR (`^`), NOT (`~`), geser kiri (`<<`), dan geser kanan (`>>`).

### F. Operator Keanggotaan dan Identitas
Operator keanggotaan (`in`, `not in`) digunakan untuk memeriksa apakah suatu nilai terdapat dalam suatu koleksi (seperti list). Operator identitas (`is`, `is not`) digunakan untuk memeriksa apakah dua variabel merujuk pada objek yang sama di memori, berbeda dengan `==` yang hanya membandingkan nilai.

## 2.2 Kode Program

### 2.2.1 Program Aritmatika_2611533015


```python
# Buat File dengan nama aritmatika_NIM.py
# Buat program untuk operator aritmatika dalam python
# Nama variabale di tambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer

angka1_3015 = int(input("Input angka-1: "))
angka2_3015 = int(input("Input angka-2: "))

# Penjumlahan
hasil_3015 = angka1_3015 + angka2_3015
print("\nOperator Penjumlahan")
print("Hasil =", hasil_3015)

# Pengurangan
hasil_3015 = angka1_3015 - angka2_3015
print("\nOperator Pengurangan")
print("Hasil =", hasil_3015)

# Perkalian
hasil_3015 = angka1_3015 * angka2_3015
print("\nOperator Perkalian")
print("Hasil =", hasil_3015)

# Pembagian, pembagian bulat, dan sisa bagi
if angka2_3015 != 0:
    hasil_3015 = angka1_3015 / angka2_3015
    print("\nOperator Pembagian")
    print("Hasil =", hasil_3015)

    hasil_3015 = angka1_3015 // angka2_3015
    print("\nOperator Pembagian Bulat")
    print("Hasil =", hasil_3015)

    hasil_3015 = angka1_3015 % angka2_3015
    print("\nOperator Sisa Bagi")
    print("Hasil =", hasil_3015)
else:
    print("Angka kedua tidak boleh bernilai 0.")

# Pangkat
hasil_3015 = angka1_3015 ** angka2_3015
print("\nOperator Pangkat")
print("Hasil =", hasil_3015)
```

**Penjelasan Program Aritmatika_2611533015:**

Program ini mendemonstrasikan **operator aritmatika** dalam Python menggunakan dua nilai yang dimasukkan pengguna.

1. **`angka1_3015 = int(input(...))` dan `angka2_3015 = int(input(...))`** — Menerima dua bilangan dari pengguna melalui `input()`, kemudian langsung dikonversi menjadi tipe data integer menggunakan `int()` agar bisa dipakai dalam perhitungan matematis.
2. **`hasil_3015 = angka1_3015 + angka2_3015`** — Operator `+` menjumlahkan kedua nilai.
3. **`hasil_3015 = angka1_3015 - angka2_3015`** — Operator `-` mengurangkan angka2 dari angka1.
4. **`hasil_3015 = angka1_3015 * angka2_3015`** — Operator `*` mengalikan kedua nilai.
5. **`if angka2_3015 != 0:`** — Kondisi ini digunakan untuk mencegah error pembagian dengan nol (`ZeroDivisionError`) sebelum melakukan operasi pembagian.
6. **`angka1_3015 / angka2_3015`** — Operator `/` menghasilkan pembagian biasa dengan hasil bertipe float.
7. **`angka1_3015 // angka2_3015`** — Operator `//` (floor division) menghasilkan hasil bagi yang dibulatkan ke bawah, berupa bilangan bulat.
8. **`angka1_3015 % angka2_3015`** — Operator `%` (modulus) menghasilkan sisa dari pembagian kedua bilangan.
9. **`hasil_3015 = angka1_3015 ** angka2_3015`** — Operator `**` menghitung angka1 dipangkatkan dengan angka2, dan baris ini tetap dijalankan di luar blok `if` karena perpangkatan tidak berisiko error meskipun angka2 bernilai 0.

Program ini menunjukkan bahwa operator aritmatika Python mencakup operasi dasar (`+`, `-`, `*`) hingga operasi khusus (`/`, `//`, `%`, `**`) yang masing-masing memiliki hasil dan kegunaan berbeda.

### 2.2.2 Program Assignment_2611533015


```python
# Buat file dengan nama assignment_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator assignment dalam Python

angka1_3015 = int(input("Input angka-1: "))
angka2_3015 = int(input("Input angka-2: "))

print("\nNilai awal angka1 =", angka1_3015)
print("Nilai angka2 =", angka2_3015)

# Assignment biasa
hasil_3015 = angka1_3015
print("\nAssignment biasa (=)")
print("Hasil =", hasil_3015)

# Assignment penambahan
hasil_3015 = angka1_3015
hasil_3015 += angka2_3015
print("\nAssignment penambahan (+=)")
print("Hasil =", hasil_3015)

# Assignment pengurangan
hasil_3015 = angka1_3015
hasil_3015 -= angka2_3015
print("\nAssignment pengurangan (-=)")
print("Hasil =", hasil_3015)

# Assignment perkalian
hasil_3015 = angka1_3015
hasil_3015 *= angka2_3015
print("\nAssignment perkalian (*=)")
print("Hasil =", hasil_3015)

# Assignment pembagian, pembagian bulat, dan sisa bagi
if angka2_3015 != 0:
    hasil_3015 = angka1_3015
    hasil_3015 /= angka2_3015
    print("\nAssignment pembagian (/=)")
    print("Hasil =", hasil_3015)

    # Operator tambahan
    hasil_3015 = angka1_3015
    hasil_3015 //= angka2_3015
    print("\nAssignment pembagian bulat (//=)")
    print("Hasil =", hasil_3015)

    hasil_3015 = angka1_3015
    hasil_3015 %= angka2_3015
    print("\nAssignment sisa bagi (%=)")
    print("Hasil =", hasil_3015)
else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka kedua tidak boleh bernilai 0.")

# Operator tambahan: assignment perpangkatan
hasil_3015 = angka1_3015
hasil_3015 **= angka2_3015
print("\nAssignment perpangkatan (**=)")
print("Hasil =", hasil_3015)
```

**Penjelasan Program Assignment_2611533015:**

Program ini mendemonstrasikan **operator assignment** dalam Python, yaitu operator yang digunakan untuk memberikan dan memperbarui nilai suatu variabel.

1. **`hasil_3015 = angka1_3015`** — Contoh assignment biasa (`=`), yaitu memberikan nilai angka1 langsung ke variabel hasil.
2. **`hasil_3015 += angka2_3015`** — Operator `+=` merupakan bentuk singkat dari `hasil_3015 = hasil_3015 + angka2_3015`, yaitu menambahkan nilai angka2 ke nilai hasil yang sudah ada.
3. **`hasil_3015 -= angka2_3015`** — Operator `-=` merupakan bentuk singkat dari pengurangan, yaitu `hasil_3015 = hasil_3015 - angka2_3015`.
4. **`hasil_3015 *= angka2_3015`** — Operator `*=` merupakan bentuk singkat dari perkalian, yaitu `hasil_3015 = hasil_3015 * angka2_3015`.
5. **`if angka2_3015 != 0:`** — Sama seperti pada program aritmatika, pengecekan ini mencegah error pembagian dengan nol sebelum operator `/=`, `//=`, dan `%=` dijalankan.
6. **`hasil_3015 /= angka2_3015`, `hasil_3015 //= angka2_3015`, `hasil_3015 %= angka2_3015`** — Masing-masing merupakan bentuk singkat dari pembagian biasa, pembagian bulat, dan sisa bagi yang langsung disimpan kembali ke variabel hasil.
7. **`hasil_3015 **= angka2_3015`** — Operator `**=` merupakan bentuk singkat dari perpangkatan, yaitu `hasil_3015 = hasil_3015 ** angka2_3015`.

Program ini menunjukkan bahwa operator assignment gabungan (`+=`, `-=`, `*=`, `/=`, `//=`, `%=`, `**=`) membuat kode lebih ringkas karena menggabungkan operasi aritmatika dan proses assignment dalam satu baris.

### 2.2.3 Program Perbandingan_2611533015


```python
# Buat file dengan nama perbandingan_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator perbandingan dalam Python

angka1_3015 = int(input("Input angka-1: "))
angka2_3015 = int(input("Input angka-2: "))

# Lebih besar dari
hasil_3015 = angka1_3015 > angka2_3015
print("\nOperator lebih besar dari")
print("angka1 > angka2 =", hasil_3015)

# Lebih kecil dari
hasil_3015 = angka1_3015 < angka2_3015
print("\nOperator lebih kecil dari")
print("angka1 < angka2 =", hasil_3015)

# Lebih besar dari atau sama dengan
hasil_3015 = angka1_3015 >= angka2_3015
print("\nOperator lebih besar dari atau sama dengan")
print("angka1 >= angka2 =", hasil_3015)

# Lebih kecil dari atau sama dengan
hasil_3015 = angka1_3015 <= angka2_3015
print("\nOperator lebih kecil dari atau sama dengan")
print("angka1 <= angka2 =", hasil_3015)

# Sama dengan
hasil_3015 = angka1_3015 == angka2_3015
print("\nOperator sama dengan")
print("angka1 == angka2 =", hasil_3015)

# Tidak sama dengan
hasil_3015 = angka1_3015 != angka2_3015
print("\nOperator tidak sama dengan")
print("angka1 != angka2 =", hasil_3015)

# Tambahan: perbandingan berantai dalam Python
hasil_3015 = 0 < angka1_3015 < 100
print("\nPerbandingan berantai")
print("0 < angka1 < 100 =", hasil_3015)

hasil_3015 = 0 < angka2_3015 < 100
print("0 < angka2 < 100 =", hasil_3015)
```

**Penjelasan Program Perbandingan_2611533015:**

Program ini mendemonstrasikan **operator perbandingan** dalam Python, yaitu operator yang menghasilkan nilai boolean (True/False) dari perbandingan dua nilai.

1. **`angka1_3015 > angka2_3015`** — Operator `>` menghasilkan True jika angka1 lebih besar dari angka2.
2. **`angka1_3015 < angka2_3015`** — Operator `<` menghasilkan True jika angka1 lebih kecil dari angka2.
3. **`angka1_3015 >= angka2_3015`** — Operator `>=` menghasilkan True jika angka1 lebih besar dari atau sama dengan angka2.
4. **`angka1_3015 <= angka2_3015`** — Operator `<=` menghasilkan True jika angka1 lebih kecil dari atau sama dengan angka2.
5. **`angka1_3015 == angka2_3015`** — Operator `==` menghasilkan True jika kedua nilai sama persis. Perlu diperhatikan bahwa `==` berbeda dengan `=` yang digunakan untuk assignment.
6. **`angka1_3015 != angka2_3015`** — Operator `!=` menghasilkan True jika kedua nilai tidak sama.
7. **`hasil_3015 = 0 < angka1_3015 < 100`** — Baris ini menunjukkan fitur **perbandingan berantai** (chained comparison) yang menjadi keunikan Python. Ekspresi ini setara dengan `(0 < angka1_3015) and (angka1_3015 < 100)`, sehingga hanya bernilai True jika angka1 berada di antara 0 dan 100.

Program ini menunjukkan bahwa setiap operator perbandingan selalu menghasilkan nilai bertipe boolean, dan Python memiliki fitur tambahan berupa perbandingan berantai yang membuat penulisan kondisi rentang nilai menjadi lebih ringkas.

### 2.2.4 Program Logika_2611533015


```python
# Buat file dengan nama logika_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: a1_1234
# Program ini menggunakan fungsi input()
# Program operator logika dalam Python

# Memasukkan nilai boolean
# Input tidak peka terhadap huruf besar dan kecil
a1_3015 = input("Input nilai boolean-1 (true/false): ").strip().lower() == "true"
a2_3015 = input("Input nilai boolean-2 (true/false): ").strip().lower() == "true"

print("\nA1 =", a1_3015)
print("A2 =", a2_3015)

# Konjungsi: bernilai True jika keduanya True
hasil_3015 = a1_3015 and a2_3015
print("\nKonjungsi (AND)")
print("A1 and A2 =", hasil_3015)

# Disjungsi: bernilai True jika salah satunya True
hasil_3015 = a1_3015 or a2_3015
print("\nDisjungsi (OR)")
print("A1 or A2 =", hasil_3015)

# Negasi A1: membalik nilai A1
hasil_3015 = not a1_3015
print("\nNegasi A1 (NOT)")
print("not A1 =", hasil_3015)

# Negasi A2: membalik nilai A2
hasil_3015 = not a2_3015
print("\nNegasi A2 (NOT)")
print("not A2 =", hasil_3015)

# XOR: bernilai True jika kedua nilai berbeda
hasil_3015 = a1_3015 != a2_3015
print("\nDisjungsi Eksklusif (XOR)")
print("A1 XOR A2 =", hasil_3015)
```

**Penjelasan Program Logika_2611533015:**

Program ini mendemonstrasikan **operator logika** dalam Python yang digunakan untuk menggabungkan atau membalik nilai boolean.

1. **`a1_3015 = input(...).strip().lower() == "true"`** — Baris ini membaca input teks dari pengguna, menghapus spasi di awal/akhir dengan `.strip()`, mengubahnya menjadi huruf kecil dengan `.lower()`, lalu membandingkannya dengan string `"true"`. Hasil perbandingan `==` inilah yang bertipe boolean dan disimpan ke a1_3015, sehingga input pengguna seperti "True", "TRUE", atau "true" tetap dikenali sama.
2. **`hasil_3015 = a1_3015 and a2_3015`** — Operator `and` (konjungsi) menghasilkan True hanya jika **kedua** operand bernilai True.
3. **`hasil_3015 = a1_3015 or a2_3015`** — Operator `or` (disjungsi) menghasilkan True jika **salah satu atau kedua** operand bernilai True.
4. **`hasil_3015 = not a1_3015`** dan **`hasil_3015 = not a2_3015`** — Operator `not` (negasi) membalik nilai boolean, dari True menjadi False atau sebaliknya.
5. **`hasil_3015 = a1_3015 != a2_3015`** — Baris ini mendemonstrasikan logika **XOR (disjungsi eksklusif)**, yang bernilai True hanya jika kedua nilai **berbeda**. Python tidak memiliki operator XOR khusus untuk boolean seperti `and`/`or`, sehingga operator `!=` dimanfaatkan karena secara logis menghasilkan perilaku yang sama dengan XOR pada nilai boolean.

Program ini menunjukkan bagaimana operator logika `and`, `or`, dan `not` digunakan untuk menggabungkan beberapa kondisi menjadi satu ekspresi boolean.

### 2.2.5 Program Bitwise_2611533015


```python
# Buat file dengan nama bitwise_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()

print("\n================================")
print("3. OPERATOR BITWISE")
print("================================")

angka1_3015 = int(input("Masukkan angka bitwise-1: "))
angka2_3015 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1 =", angka1_3015, "| biner =", bin(angka1_3015))
print("angka2 =", angka2_3015, "| biner =", bin(angka2_3015))

# Bitwise AND
hasil_3015 = angka1_3015 & angka2_3015
print("\nBitwise AND (&)")
print(angka1_3015, "&", angka2_3015, "=", hasil_3015)
print("Biner hasil =", bin(hasil_3015))
print("Biner hasil (8 bit) =", format(hasil_3015, "08b"))

# Bitwise OR
hasil_3015 = angka1_3015 | angka2_3015
print("\nBitwise OR (|)")
print(angka1_3015, "|", angka2_3015, "=", hasil_3015)
print("Biner hasil =", bin(hasil_3015))
print("Biner hasil (8 bit) =", format(hasil_3015, "08b"))

# Bitwise XOR
hasil_3015 = angka1_3015 ^ angka2_3015
print("\nBitwise XOR (^)")
print(angka1_3015, "^", angka2_3015, "=", hasil_3015)
print("Biner hasil =", bin(hasil_3015))
print("Biner hasil (8 bit) =", format(hasil_3015, "08b"))

# Bitwise NOT
hasil_3015 = ~angka1_3015
print("\nBitwise NOT (~)")
print("~", angka1_3015, "=", hasil_3015)
print("Biner hasil =", bin(hasil_3015))
print("Biner hasil (8 bit) =", format(hasil_3015, "08b"))

# Bitwise geser kiri
jumlah_geser_3015 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_3015 = angka1_3015 << jumlah_geser_3015
print("\nBitwise geser kiri (<<)")
print(angka1_3015, "<<", jumlah_geser_3015, "=", hasil_3015)
print("Biner hasil =", bin(hasil_3015))
print("Biner hasil (8 bit) =", format(hasil_3015, "08b"))

# Bitwise geser kanan
hasil_3015 = angka1_3015 >> jumlah_geser_3015
print("\nBitwise geser kanan (>>)")
print(angka1_3015, ">>", jumlah_geser_3015, "=", hasil_3015)
print("Biner hasil =", bin(hasil_3015))
print("Biner hasil (8 bit) =", format(hasil_3015, "08b"))
```

**Penjelasan Program Bitwise_2611533015:**

Program ini mendemonstrasikan **operator bitwise** dalam Python, yaitu operator yang bekerja pada representasi biner dari suatu bilangan.

1. **`bin(angka1_3015)`** — Fungsi `bin()` mengubah bilangan desimal menjadi representasi biner (dengan awalan `0b`), digunakan agar hasil operasi bitwise dapat dibandingkan langsung dalam bentuk bit.
2. **`hasil_3015 = angka1_3015 & angka2_3015`** — Operator `&` (AND) membandingkan setiap bit; hasilnya 1 hanya jika kedua bit pada posisi yang sama bernilai 1.
3. **`hasil_3015 = angka1_3015 | angka2_3015`** — Operator `|` (OR) menghasilkan bit 1 jika salah satu atau kedua bit pada posisi yang sama bernilai 1.
4. **`hasil_3015 = angka1_3015 ^ angka2_3015`** — Operator `^` (XOR) menghasilkan bit 1 hanya jika kedua bit pada posisi yang sama **berbeda**.
5. **`hasil_3015 = ~angka1_3015`** — Operator `~` (NOT) membalik seluruh bit dari angka1. Pada Python, hasil `~x` secara matematis setara dengan `-(x + 1)` karena bilangan bulat direpresentasikan dalam bentuk komplemen dua (two's complement), sehingga hasilnya bisa berupa bilangan negatif.
6. **`hasil_3015 = angka1_3015 << jumlah_geser_3015`** — Operator `<<` (geser kiri) menggeser seluruh bit angka1 ke kiri sebanyak jumlah_geser posisi, yang secara matematis setara dengan mengalikan angka1 dengan 2 pangkat jumlah_geser.
7. **`hasil_3015 = angka1_3015 >> jumlah_geser_3015`** — Operator `>>` (geser kanan) menggeser seluruh bit angka1 ke kanan sebanyak jumlah_geser posisi, setara dengan pembagian bulat angka1 dengan 2 pangkat jumlah_geser.
8. **`format(hasil_3015, "08b")`** — Digunakan untuk menampilkan hasil dalam format biner 8 digit (diberi padding angka 0 di depan) agar lebih mudah dibaca dan dibandingkan antar bit.

Program ini menunjukkan bagaimana operator bitwise memanipulasi data pada level bit, yang sering digunakan dalam pemrograman sistem, enkripsi sederhana, dan optimasi perhitungan.

### 2.2.6 Program Lainnya (Keanggotaan & Identitas)_2611533015


```python
# Buat file dengan nama lainnya_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Program operator keanggotaan dan identitas

print("================================")
print("1. OPERATOR KEANGGOTAAN")
print("================================")

# Input beberapa data yang dipisahkan dengan koma
input_data_3015 = input("Masukkan beberapa angka, pisahkan dengan koma: ")

# Mengubah input menjadi list integer
data_3015 = [int(angka.strip()) for angka in input_data_3015.split(",")]

nilai_dicari_3015 = int(input("Masukkan angka yang ingin dicari: "))

# Operator in
hasil_3015 = nilai_dicari_3015 in data_3015
print("\nOperator keanggotaan IN")
print(nilai_dicari_3015, "in", data_3015, "=", hasil_3015)

# Operator not in
hasil_3015 = nilai_dicari_3015 not in data_3015
print("\nOperator keanggotaan NOT IN")
print(nilai_dicari_3015, "not in", data_3015, "=", hasil_3015)

print("\n================================")
print("2. OPERATOR IDENTITAS")
print("================================")

# objek1 menggunakan list dari input pengguna
objek1_3015 = data_3015

# objek2 merujuk pada objek yang sama dengan objek1
objek2_3015 = objek1_3015

# objek3 memiliki isi sama, tetapi merupakan objek baru
objek3_3015 = data_3015.copy()

print("objek1 =", objek1_3015)
print("objek2 =", objek2_3015)
print("objek3 =", objek3_3015)

# Operator is
hasil_3015 = objek1_3015 is objek2_3015
print("\nOperator identitas IS")
print("objek1 is objek2 =", hasil_3015)

# Operator is not
hasil_3015 = objek1_3015 is not objek3_3015
print("\nOperator identitas IS NOT")
print("objek1 is not objek3 =", hasil_3015)

# Membandingkan identitas dan nilai
print("\nPerbandingan identitas dan nilai")
print("objek1 is objek3 =", objek1_3015 is objek3_3015)
print("objek1 == objek3 =", objek1_3015 == objek3_3015)
```

Program ini mendemonstrasikan **operator keanggotaan** (`in`, `not in`) dan **operator identitas** (`is`, `is not`) dalam Python.

**Bagian 1 — Operator Keanggotaan:**

**Bagian 2 — Operator Identitas:**
4. **`objek2_3015 = objek1_3015`** — Baris ini tidak membuat list baru, melainkan hanya membuat objek2 **merujuk ke objek yang sama persis** dengan objek1 di memori.

6. **`hasil_3015 = objek1_3015 is objek2_3015`** — Operator `is` menghasilkan True karena objek1 dan objek2 merujuk ke **alamat memori yang sama**.

8. **`objek1_3015 is objek3_3015`** dibandingkan dengan **`objek1_3015 == objek3_3015`** — Baris terakhir ini menegaskan perbedaan mendasar antara `is` dan `==`: `is` bernilai False karena membandingkan identitas/alamat memori (objek1 dan objek3 berbeda objek), sedangkan `==` bernilai True karena membandingkan isi/nilai (kedua list memiliki elemen yang sama).

Program ini menunjukkan bahwa operator keanggotaan berguna untuk memeriksa keberadaan nilai dalam koleksi, sedangkan operator identitas berguna untuk membedakan antara dua variabel yang merujuk objek yang sama dengan dua variabel yang nilainya kebetulan sama tetapi merupakan objek berbeda.

Berdasarkan praktikum Operator, dapat disimpulkan:

1. **Operator aritmatika** (`+`, `-`, `*`, `/`, `//`, `%`, `**`) digunakan untuk melakukan perhitungan matematis dalam Python.
2. **Operator assignment** (`=`, `+=`, `-=`, `*=`, `/=`, `//=`, `%=`, `**=`) digunakan untuk memberikan dan memperbarui nilai variabel secara ringkas.
3. **Operator perbandingan** (`>`, `<`, `>=`, `<=`, `==`, `!=`) menghasilkan nilai boolean dari perbandingan dua nilai, termasuk fitur perbandingan berantai.
4. **Operator logika** (`and`, `or`, `not`) digunakan untuk menggabungkan atau membalik beberapa ekspresi boolean.
5. **Operator bitwise** (`&`, `|`, `^`, `~`, `<<`, `>>`) digunakan untuk memanipulasi data pada level bit/biner.
6. **Operator keanggotaan** (`in`, `not in`) memeriksa keberadaan nilai dalam koleksi, sedangkan **operator identitas** (`is`, `is not`) memeriksa apakah dua variabel merujuk objek yang sama di memori — berbeda dengan `==` yang membandingkan nilai.
7. Pemahaman mengenai berbagai jenis operator adalah fondasi penting untuk menyusun ekspresi, kondisi, dan logika program yang lebih kompleks.

```python
# ============================================================
#  SISTEM SIMULASI TRANSAKSI DAN VALIDASI AKSES TOKO
#  Tugas Praktikum 3 - Operator Python
#  NIM 2611533015
#
#  Program kasir sederhana untuk menghitung total transaksi,
#  memvalidasi status pelanggan, dan menentukan hak akses promo.
#  Seluruh jenis operator Python terintegrasi dalam satu alur.
# ============================================================

print("=== SISTEM TRANSAKSI TOKO ===\n")

# ---------- INPUT DATA PELANGGAN ----------
nama_3015 = input("Masukkan Nama Pelanggan : ").strip()
status_3015 = input("Masukkan Status Pelanggan (member/nonmember) : ").strip().lower()
total_belanja_3015 = float(input("Masukkan Total Belanja : "))
jumlah_barang_3015 = int(input("Masukkan Jumlah Barang : "))
kode_promo_3015 = input("Masukkan Kode Promo : ").strip().upper()

# daftar promo
daftar_promo_3015 = ["HEMAT10", "HEMAT20", "GRATISONGKIR"]

# ---------- OPERATOR PERBANDINGAN ----------
# Memeriksa apakah syarat diskon / promo terpenuhi (hasil True/False)
minimal_belanja_3015 = total_belanja_3015 >= 200000
minimal_barang_3015 = jumlah_barang_3015 >= 3
status_member_3015 = status_3015 == "member"
bukan_member_3015 = status_3015 != "member"

# ---------- OPERATOR LOGIKA ----------
# Diskon khusus member : member AND belanja mencapai batas minimum
diskon_member_3015 = status_member_3015 and minimal_belanja_3015
# Syarat mendapat poin bonus : member OR jumlah barang minimal 3
syarat_poin_3015 = status_member_3015 or minimal_barang_3015
# Kebalikan dari status member
nonmember_3015 = not status_member_3015

# ---------- OPERATOR KEANGGOTAAN ----------
# Memeriksa apakah kode promo user terdapat dalam daftar promo
promo_tersedia_3015 = kode_promo_3015 in daftar_promo_3015
promo_tidak_ada_3015 = kode_promo_3015 not in daftar_promo_3015
# Pelanggan mendapatkan promo jika jumlah barang >= 3 DAN promo tersedia
dapat_promo_3015 = minimal_barang_3015 and promo_tersedia_3015

# ---------- OPERATOR ARITMATIKA DAN PENUGASAN ----------
# Besar diskon dihitung bertahap memakai augmented assignment (+=)
besar_diskon_3015 = 0.0                     # operator penugasan (=)
if minimal_belanja_3015:                    # 10% utk belanja >= Rp200.000
    besar_diskon_3015 += (10.0 / 100) * total_belanja_3015
if diskon_member_3015:                      # +5% khusus member
    besar_diskon_3015 += (5.0 / 100) * total_belanja_3015
if dapat_promo_3015:                        # +5% bila kode promo valid
    besar_diskon_3015 += (5.0 / 100) * total_belanja_3015

# Pelanggan dikatakan mendapat diskon bila besar diskon lebih dari 0
mendapat_diskon_3015 = besar_diskon_3015 > 0

# Total pembayaran setelah dipotong diskon (operator pengurangan)
total_bayar_3015 = total_belanja_3015       # penugasan awal
total_bayar_3015 -= besar_diskon_3015       # augmented assignment (-=)

# Harga rata-rata barang setelah diskon (operator pembagian)
if jumlah_barang_3015 > 0:
    rata_rata_barang_3015 = total_bayar_3015 / jumlah_barang_3015
else:
    rata_rata_barang_3015 = 0.0

# Sisa pembagian total belanja bila dibayar pecahan Rp10.000 (modulo %)
sisa_pembagian_3015 = int(total_belanja_3015) % 10000

# Poin loyalitas : 10 poin per Rp50.000, dikali 2 bila syarat terpenuhi
poin_3015 = 0
poin_3015 += int(total_bayar_3015 // 50000) * 10
if syarat_poin_3015:
    poin_3015 *= 2                          # augmented assignment (*=)

# ---------- OPERATOR IDENTITAS ----------
# is  -> membandingkan IDENTITAS objek (apakah menunjuk objek yang sama)
# ==  -> membandingkan NILAI/ISI objek (apakah isinya sama)
referensi_promo_3015 = daftar_promo_3015            # objek yang SAMA
salinan_promo_3015 = list(daftar_promo_3015)        # objek BARU, isi sama

identitas_sama_3015 = daftar_promo_3015 is referensi_promo_3015
identitas_beda_3015 = daftar_promo_3015 is not salinan_promo_3015
nilai_sama_3015 = daftar_promo_3015 == salinan_promo_3015

# ---------- OPERATOR BITWISE ----------
# Setiap kondisi direpresentasikan dalam bit biner :
#   0001 = member         0100 = jumlah barang >= 3
#   0010 = belanja >= 200rb   1000 = kode promo tersedia
bit_member_3015 = 0b0001 if status_member_3015 else 0
bit_belanja_3015 = 0b0010 if minimal_belanja_3015 else 0
bit_barang_3015 = 0b0100 if minimal_barang_3015 else 0
bit_promo_3015 = 0b1000 if promo_tersedia_3015 else 0

# OR (|) : menggabungkan seluruh kondisi menjadi satu kode status
kode_status_3015 = (bit_member_3015 | bit_belanja_3015 |
                    bit_barang_3015 | bit_promo_3015)

# AND (&) : memeriksa apakah suatu kondisi terpenuhi dalam kode status
cek_akses_member_3015 = kode_status_3015 & 0b0001
cek_akses_promo_3015 = kode_status_3015 & 0b1000

# Kode referensi transaksi lain : member + belanja minimal + barang minimal
kode_referensi_3015 = 0b1011

# XOR (^) : membandingkan perbedaan status antara dua kode transaksi
perbedaan_status_3015 = kode_status_3015 ^ kode_referensi_3015

# Shift kiri (<<) : menggeser kode status untuk keperluan cadangan
kode_shift_3015 = kode_status_3015 << 1

# ---------- HAK AKSES PELANGGAN ----------
hak_member_3015 = status_member_3015
hak_promo_3015 = dapat_promo_3015
hak_gratis_ongkir_3015 = kode_promo_3015 == "GRATISONGKIR"

# ================= TAMPILAN HASIL =================
print("\n=== DATA TRANSAKSI ===")
print(f"Nama Pelanggan     : {nama_3015}")
print(f"Status Pelanggan   : {status_3015}")
print(f"Total Belanja      : Rp{total_belanja_3015:.0f}")
print(f"Jumlah Barang      : {jumlah_barang_3015}")
print(f"Kode Promo         : {kode_promo_3015}")

print("\n=== HASIL VALIDASI ===")
print(f"Belanja >= Rp200.000     : {minimal_belanja_3015}")
print(f"Jumlah Barang >= 3       : {minimal_barang_3015}")
print(f"Status Member            : {status_member_3015}")
print(f"Kode Promo Tersedia      : {promo_tersedia_3015}")
print(f"Mendapatkan Diskon       : {mendapat_diskon_3015}")
print(f"Mendapatkan Promo        : {dapat_promo_3015}")

print("\n=== HASIL PERHITUNGAN ===")
print(f"Besar Diskon             : Rp{besar_diskon_3015:.0f}")
print(f"Total Pembayaran         : Rp{total_bayar_3015:.0f}")
print(f"Rata-rata Harga Barang   : Rp{rata_rata_barang_3015:.0f}")
print(f"Sisa Pembagian (modulo)  : Rp{sisa_pembagian_3015}")
print(f"Poin Loyalitas           : {poin_3015} poin")

print("\n=== HAK AKSES PELANGGAN ===")
print(f"Member Access          : {hak_member_3015}")
print(f"Promo Access           : {hak_promo_3015}")
print(f"Free Shipping Access   : {hak_gratis_ongkir_3015}")

print("\n=== HASIL OPERATOR ===")
print("\n--- OPERATOR ARITMATIKA ---")
print(f"Total belanja                        : Rp{total_belanja_3015:.0f}")
print(f"Diskon                                : Rp{besar_diskon_3015:.0f}")
print(f"Total bayar (total belanja - diskon)  : Rp{total_bayar_3015:.0f}")
print(f"Rata-rata (total bayar / jumlah barang) : Rp{rata_rata_barang_3015:.0f}")
print(f"Modulo (total % 10000)                : {sisa_pembagian_3015}")

print("\n--- OPERATOR PERBANDINGAN ---")
print(f"{total_belanja_3015:.0f} >= 200000  -> {minimal_belanja_3015}")
print(f"{jumlah_barang_3015} >= 3         -> {minimal_barang_3015}")
print(f"status == 'member'                  -> {status_member_3015}")

print("\n--- OPERATOR LOGIKA ---")
print(f"member AND belanja_min              -> {diskon_member_3015}")
print(f"member OR  barang >= 3              -> {syarat_poin_3015}")
print(f"NOT member                          -> {nonmember_3015}")

print("\n--- OPERATOR PENUGASAN ---")
print(f"besar_diskon dihitung dengan +=     -> Rp{besar_diskon_3015:.0f}")
print(f"total_bayar dikurangi dengan -=     -> Rp{total_bayar_3015:.0f}")
print(f"poin digandakan dengan *=           -> {poin_3015} poin")

print("\n--- OPERATOR KEANGGOTAAN ---")
print(f"Daftar promo               : {daftar_promo_3015}")
print(f"'{kode_promo_3015}' in daftar      -> {promo_tersedia_3015}")
print(f"'{kode_promo_3015}' not in daftar  -> {promo_tidak_ada_3015}")

print("\n--- OPERATOR IDENTITAS ---")
print(f"daftar is referensi (objek sama)   -> {identitas_sama_3015}")
print(f"daftar is not salinan (objek baru) -> {identitas_beda_3015}")
print(f"daftar == salinan (nilai sama)     -> {nilai_sama_3015}")
print("Catatan : is membandingkan identitas objek, == membandingkan nilai/isi objek.")

print("\n--- OPERATOR BITWISE ---")
print("Format bit : 0001=member | 0010=belanja>=200rb | 0100=barang>=3 | 1000=promo")
print(f"OR (|) : {format(bit_member_3015, '04b')} | {format(bit_belanja_3015, '04b')} | {format(bit_barang_3015, '04b')} | {format(bit_promo_3015, '04b')} = {format(kode_status_3015, '04b')}")
print(f"Kode Status Transaksi   : {format(kode_status_3015, '04b')} (desimal {kode_status_3015})")
print(f"AND (&) cek member      : {format(kode_status_3015, '04b')} & 0001 = {format(cek_akses_member_3015, '04b')} ({cek_akses_member_3015})")
print(f"AND (&) cek promo       : {format(kode_status_3015, '04b')} & 1000 = {format(cek_akses_promo_3015, '04b')} ({cek_akses_promo_3015})")
print(f"XOR (^) beda status     : {format(kode_status_3015, '04b')} ^ {format(kode_referensi_3015, '04b')} = {format(perbedaan_status_3015, '04b')} ({perbedaan_status_3015})")
print(f"Shift (<<) kode         : {format(kode_status_3015, '04b')} << 1 = {format(kode_shift_3015, 'b')} ({kode_shift_3015})")

print("\n=== SELESAI ===")
```

**2. Jelaskan perbedaan operator and, or, dan not berdasarkan penerapannya pada program.**

Operator `and` digunakan pada `diskon_member_3015 = status_member_3015 and minimal_belanja_3015`, yang hanya bernilai True jika **kedua** syarat terpenuhi sekaligus (pelanggan member **dan** belanjanya mencapai minimum) — cocok untuk aturan yang mengharuskan semua syarat terpenuhi. Operator `or` digunakan pada `syarat_poin_3015 = status_member_3015 or minimal_barang_3015`, yang bernilai True jika **salah satu** syarat terpenuhi (member **atau** jumlah barang minimal 3) — cocok untuk aturan yang cukup satu syarat saja. Operator `not` digunakan pada `nonmember_3015 = not status_member_3015` untuk membalik nilai boolean, menghasilkan True jika pelanggan bukan member.

**3. Jelaskan fungsi operator in dan not in pada pemeriksaan kode promo.**

Operator `in` pada `promo_tersedia_3015 = kode_promo_3015 in daftar_promo_3015` memeriksa apakah kode promo yang dimasukkan pengguna **terdapat** di dalam `daftar_promo_3015` (list berisi "HEMAT10", "HEMAT20", "GRATISONGKIR"), menghasilkan True jika kode valid. Sebaliknya, operator `not in` pada `promo_tidak_ada_3015` memeriksa apakah kode promo **tidak terdapat** dalam daftar tersebut. Hasil dari `in` inilah yang menjadi salah satu syarat pada `dapat_promo_3015 = minimal_barang_3015 and promo_tersedia_3015`, sehingga promo hanya diberikan jika kodenya benar-benar terdaftar.

**4. Apa perbedaan antara is dan == pada Python?**

Operator `is` membandingkan **identitas objek**, yaitu apakah dua variabel menunjuk ke alamat memori (objek) yang sama persis, sedangkan `==` membandingkan **nilai/isi** dari objek tersebut. Pada program, `referensi_promo_3015 = daftar_promo_3015` membuat variabel yang menunjuk ke objek list yang **sama**, sehingga `daftar_promo_3015 is referensi_promo_3015` bernilai True. Sebaliknya, `salinan_promo_3015 = list(daftar_promo_3015)` membuat objek list **baru** dengan isi yang identik, sehingga `daftar_promo_3015 is not salinan_promo_3015` bernilai True (objek berbeda), tetapi `daftar_promo_3015 == salinan_promo_3015` tetap bernilai True karena isinya sama.

**5. Jelaskan bagaimana operator bitwise digunakan untuk menentukan hak akses pelanggan.**

Setiap syarat pelanggan diwakili oleh satu bit: member (`0001`), belanja ≥ Rp200.000 (`0010`), jumlah barang ≥ 3 (`0100`), dan kode promo tersedia (`1000`). Operator `OR (|)` menggabungkan keempat bit ini menjadi satu `kode_status_3015` yang merepresentasikan seluruh kondisi pelanggan sekaligus. Operator `AND (&)` kemudian digunakan untuk memeriksa apakah bit tertentu aktif dalam kode status tersebut, misalnya `kode_status_3015 & 0b0001` untuk mengecek akses member dan `kode_status_3015 & 0b1000` untuk mengecek akses promo — hasilnya bukan nol jika bit itu aktif. Operator `XOR (^)` dipakai untuk membandingkan perbedaan status antara kode transaksi pelanggan dengan kode referensi, dan operator geser kiri (`<<`) dipakai untuk menggeser kode status sebagai kebutuhan cadangan/backup kode.

**6. Jelaskan alur program mulai dari input hingga menghasilkan total pembayaran dan status promo.**

Alur program dimulai dari **input** nama, status pelanggan, total belanja, jumlah barang, dan kode promo. Selanjutnya, **operator perbandingan** memeriksa syarat-syarat dasar (belanja minimum, jumlah barang minimum, status member). Hasil perbandingan ini digabungkan dengan **operator logika** untuk menentukan kelayakan diskon member dan syarat poin, sementara **operator keanggotaan** memeriksa validitas kode promo terhadap daftar promo. Berdasarkan kombinasi syarat tersebut, **operator aritmatika dan penugasan** (`+=`) menghitung `besar_diskon_3015` secara bertahap, lalu `total_bayar_3015` diperoleh dengan mengurangkan (`-=`) diskon dari total belanja. Nilai `total_bayar_3015` ini juga dipakai untuk menghitung rata-rata harga barang dan poin loyalitas. Di sisi lain, syarat-syarat yang sama diubah menjadi bit dan digabungkan dengan **operator bitwise** untuk membentuk `kode_status_3015`, yang menjadi dasar penentuan hak akses member, promo, dan gratis ongkir. Seluruh hasil akhirnya ditampilkan melalui rangkaian `print()`.

**7. Identifikasi minimal satu kondisi dalam program yang menghasilkan True dan satu kondisi yang menghasilkan False.**

Contohnya, jika pengguna memasukkan total belanja Rp250.000 (lebih besar dari Rp200.000), maka `minimal_belanja_3015 = total_belanja_3015 >= 200000` menghasilkan **True**. Sebaliknya, jika status pelanggan yang diinput adalah "nonmember", maka `status_member_3015 = status_3015 == "member"` menghasilkan **False**, karena isi variabel `status_3015` ("nonmember") tidak sama dengan string "member". Kedua kondisi ini kemudian memengaruhi hasil operator logika dan bitwise selanjutnya dalam program.
