## 2.1 Dasar Teori

Praktikum Perulangan For membahas berbagai jenis penerapan perulangan for dalam Python serta cara penggunaannya untuk membentuk pola karakter.

### A. Perulangan For Dasar
Perulangan `for` digunakan untuk mengulang sekumpulan pernyataan pada setiap elemen di dalam suatu objek iterable (benda yang dapat diulang), misalnya hasil fungsi `range()`. Sintaks dasarnya adalah:

```
for variabel in iterable:
    pernyataan_yang_diulang
```

Pada setiap iterasi, variabel akan berisi satu nilai dari iterable dan blok pernyataan di dalamnya dijalankan.

### B. Fungsi range()
Fungsi `range()` menghasilkan urutan bilangan bulat. Bentuk umumnya:
- `range(stop)` — menghasilkan bilangan 0 sampai `stop-1`.
- `range(start, stop)` — menghasilkan bilangan `start` sampai `stop-1`.
- `range(start, stop, step)` — menghasilkan bilangan `start` sampai `stop-1` dengan kenaikan `step` (dapat bernilai negatif untuk menghitung mundur).

### C. Perulangan Bersarang (Nested For)
Nested for adalah perulangan for di dalam perulangan for. Pada setiap iterasi perulangan luar (outer), perulangan dalam (inner) dijalankan secara penuh dari awal hingga akhir. Teknik ini sangat berguna untuk membentuk pola dua dimensi seperti matriks, segitiga, dan bangun geometri karakter.

### D. Akumulasi Nilai pada Perulangan
Variabel akumulator (misalnya `jumlah_3015`) digunakan untuk mengumpulkan hasil pada setiap iterasi, contoh: `jumlah_3015 = jumlah_3015 + i`. Setelah perulangan selesai, variabel tersebut berisi total seluruh nilai yang diakumulasikan.

### E. F-String dan Parameter end pada print()
F-string (`print(f"teks {variabel}")`) digunakan untuk menyisipkan nilai variabel ke dalam teks keluaran, sedangkan parameter `end=""` pada `print()` digunakan agar keluaran berikutnya tetap berada pada baris yang sama, sehingga memungkinkan pembentukan pola karakter baris demi baris.

Seluruh program pada praktikum ini disusun menggunakan pilar-pilar perulangan for tersebut; sebagian program dilengkapi percabangan if-else dan operasi string di dalam blok perulangan untuk mengontrol karakter yang dicetak.

## 2.2 Kode Program

### 2.2.1 Program perulangan_for1_2611533015 (Perulangan For Dasar)


````python
# Buat file dengan nama perulangan_for1_NIM_py
# Buat program untuk perulangan for dalam python
# Nama variable ditabah 4 digit nim terakhir contoh: ulang_1234
# program ini menggunakan fungsi input()

ulang_3015 = int(input("Masukkan Jumblah Perulangan: "))

for i in range(ulang_3015):
    print(f"perulangan ke-{i}")
````

**Penjelasan Program perulangan_for1_2611533015:**

Program ini mendemonstrasikan **perulangan for dasar** menggunakan fungsi `range()`.

1. **`ulang_3015 = int(input("Masukkan Jumblah Perulangan: "))`** — Menerima jumlah perulangan dari pengguna (berupa string) lalu dikonversi menjadi integer menggunakan `int()`.
2. **`for i in range(ulang_3015):`** — Melakukan perulangan mulai dari `i = 0` sampai `i = ulang_3015 - 1` (sebanyak `ulang_3015` kali). Nilai `i` otomatis diisi oleh Python pada setiap iterasi.
3. **`print(f"perulangan ke-{i}")`** — Mencetak teks dengan f-string yang menyisipkan nilai `i` secara dinamis ke dalam kalimat.

Pada contoh di atas dengan input `5`, blok di dalam perulangan dijalankan 5 kali dan mencetak `perulangan ke-0` sampai `perulangan ke-4`. Program ini menunjukkan bahwa blok pernyataan di dalam `for` akan dieksekusi berulang kali sesuai jumlah yang dimasukkan pengguna.


### 2.2.2 Program perulangan_for2_2611533015 (Rentang Awal dan Akhir range())


````python
# Buat file dengan nama perulangan_for2_NIM_py
# Buat program untuk perulangan for dalam python
# Nama variable ditabah 4 digit nim terakhir contoh: ulang_1234
# program ini menggunakan fungsi input()

ulang_3015 = int(input("Masukkan Jumblah Perulangan: "))
print ("Perulangan ke-0 sampai ke-", ulang_3015-1)
for i in range(ulang_3015):
    print(i, end=" ")
print()
print("Perulangan ke-1 sampai ke-", ulang_3015)
for i in range(1,ulang_3015+1):
    print(i, end=" ")
````

**Penjelasan Program perulangan_for2_2611533015:**

Program ini menunjukkan dua bentuk pemakaian fungsi `range()` pada perulangan for.

1. **`for i in range(ulang_3015):`** — Rentang dimulai dari `0` sampai `ulang_3015 - 1`. Apabila input `5`, nilai `i` adalah `0, 1, 2, 3, 4`, sehingga dicetak "Perulangan ke-0 sampai ke- 4" lalu deret `0 1 2 3 4`.
2. **`print(i, end=" ")`** — Parameter `end=" "` menggantikan akhir baris default dengan spasi agar angka-angka tercetak pada satu baris.
3. **`print()`** — Pindah ke baris baru setelah deret pertama selesai.
4. **`for i in range(1, ulang_3015 + 1):`** — Rentang dimulai dari `1` sampai `ulang_3015` (karena nilai akhir `ulang_3015 + 1` tidak ikut dihitung). Dengan input `5`, dicetak deret `1 2 3 4 5`.

Program ini menunjukkan bahwa memilih nilai awal dan nilai akhir pada `range()` akan menentukan rentang iterasi yang dihasilkan.


### 2.2.3 Program perulangan_for3_2611533015 (Akumulasi Penjumlahan Deret)


````python
# Buat file dengan nama perulangan_for3_NIM_py
# Buat program untuk perulangan for dalam python
# Nama variable ditabah 4 digit nim terakhir contoh: ulang_1234
# program ini menggunakan fungsi input()

ulang_3015 = int(input("Masukkan Jumblah Perulangan: "))

jumlah_3015 = 0
for i in range(1, ulang_3015 + 1):
    print(i, end=" ")
    jumlah_3015 = jumlah_3015 + i

    if i < ulang_3015:
        print(" + ", end="")
    else:
        print(" = ", end="")
print()
print(f"jumlah = {jumlah_3015}")
````

**Penjelasan Program perulangan_for3_2611533015:**

Program ini mendemonstrasikan **akumulasi nilai** menggunakan perulangan for.

1. **`jumlah_3015 = 0`** — Variabel akumulator diinisialisasi dengan nilai 0 sebelum perulangan dimulai.
2. **`for i in range(1, ulang_3015 + 1):`** — Perulangan dari `i = 1` sampai `i = ulang_3015`.
3. **`print(i, end=" ")`** — Mencetak angka saat ini pada baris yang sama.
4. **`jumlah_3015 = jumlah_3015 + i`** — Nilai `i` ditambahkan ke akumulator pada setiap iterasi, sehingga setelah perulangan selesai `jumlah_3015` berisi total seluruh angka.
5. **`if i < ulang_3015:`** — Selama belum sampai angka terakhir, dicetak simbol `" + "`; pada iterasi terakhir (`else`) dicetak `" = "`. Percabangan di dalam perulangan ini mengendalikan tampilan deret penjumlahan.
6. **`print(f"jumlah = {jumlah_3015}")`** — Mencetak hasil akhir penjumlahan, yaitu `1 + 2 + 3 + 4 + 5 = 15`.

Program ini menunjukkan pola dasar akumulator yang sangat berguna untuk menghitung total, rata-rata, dan nilai agregat lainnya.


### 2.2.4 Program nested_for1_2611533015 (Nested For Pola Segitiga)


````python
# Buat file dengan nama nested_for1_NIM_py
# Buat program untuk perulangan for dalam python
# Nama variable ditabah 4 digit nim terakhir contoh: ulang_1234
# program ini menggunakan fungsi input()

batas_3015 = int(input("Masukkan Nilai Batas: "))
for line_3015 in range(1, batas_3015+1):
    for j in range(1, (-1 * line_3015 + batas_3015) + 1):
        print(" . ", end="")
    print(line_3015)
````

**Penjelasan Program nested_for1_2611533015:**

Program ini mendemonstrasikan **perulangan bersarang (nested for)** untuk mencetak pola segitiga.

1. **`for line_3015 in range(1, batas_3015 + 1):`** — Perulangan luar berjalan dari baris `1` sampai `batas_3015`.
2. **`for j in range(1, (-1 * line_3015 + batas_3015) + 1):`** — Perulangan dalam. Ekspresi `(-1 * line_3015 + batas_3015) + 1` menghitung banyaknya titik pada tiap baris; semakin besar nilai `line_3015`, semakin sedikit titik yang dicetak sehingga terbentuk pola mengecil.
3. **`print(" . ", end="")`** — Mencetak pola titik pada baris yang sama menggunakan parameter `end=""`.
4. **`print(line_3015)`** — Setelah perulangan dalam selesai, dicetak nomor baris lalu pindah ke baris berikutnya.

Dengan input `5`, baris pertama mencetak 4 titik dan angka `1`, baris kedua 3 titik dan angka `2`, dan seterusnya hingga baris kelima hanya mencetak angka `5` tanpa titik. Program ini menunjukkan cara mengontrol banyaknya iterasi perulangan dalam berdasarkan variabel perulangan luar (`line_3015`).


### 2.2.5 Program nested_for2_2611533015 (Nested For Pola Bintang)


````python
# Buat file dengan nama nested_for2_NIM_py
# Buat program untuk perulangan for dalam python
# Nama variable ditabah 4 digit nim terakhir contoh: ulang_1234
# program ini menggunakan fungsi input()

batas_3015 = int(input("Masukkan Nilai Batas: "))
for i in range(1, batas_3015+1):
    for j in range(1, batas_3015 + 1):
        print(" * ", end="")
     # pindah ke baris berikutnya
````

**Penjelasan Program nested_for2_2611533015:**

Program ini mendemonstrasikan **perulangan bersarang** yang mencetak pola bintang.

1. **`for i in range(1, batas_3015 + 1):`** — Perulangan luar; dengan input `5`, berjalan 5 kali (i = 1 s.d. 5).
2. **`for j in range(1, batas_3015 + 1):`** — Perulangan dalam yang juga berjalan `batas_3015` kali pada setiap iterasi perulangan luar, sehingga total perintah `print(" * ", end="")` dieksekusi sebanyak `batas_3015 x batas_3015` = 25 kali.
3. **`print(" * ", end="")`** — Mencetak bintang pada baris yang sama.
4. **`# pindah ke baris berikutnya`** — Hanya berupa komentar; karena tidak ada perintah `print()` pemindahan baris di dalam perulangan luar, seluruh bintang tercetak dalam satu baris yang panjang.

Agar terbentuk matriks `batas x batas`, tambahkan perintah `print()` di akhir blok perulangan luar. Program ini mengajarkan bahwa pemindahan baris perlu dilakukan secara eksplisit agar pola terbentuk sesuai keinginan.


### 2.2.6 Program nested_for3_2611533015 (Nested For Matriks Bilangan)


````python
# Buat file dengan nama nested_for3_NIM_py
# Buat program untuk perulangan for dalam python
# Nama variable ditabah 4 digit nim terakhir contoh: ulang_1234
# program ini menggunakan fungsi input()

batas_3015 = int(input("Masukkan Nilai Batas: "))
for i in range(1, batas_3015+1):
    for j in range(batas_3015 + 1):
        print(j+i, end=" ")
    print() # pindah ke baris berikutnya
````

**Penjelasan Program nested_for3_2611533015:**

Program ini mendemonstrasikan **perulangan bersarang** untuk membentuk matriks bilangan.

1. **`for i in range(1, batas_3015 + 1):`** — Perulangan luar mengatur baris, bernilai `1` sampai `batas_3015`.
2. **`for j in range(batas_3015 + 1):`** — Perulangan dalam. `range(batas_3015 + 1)` menghasilkan `j = 0, 1, 2, 3, 4, 5` (sebanyak `batas + 1` kolom).
3. **`print(j + i, end=" ")`** — Mencetak hasil penjumlahan `j + i`; karena `j` dimulai dari 0, nilai terkecil pada setiap baris adalah `i`, sehingga baris demi baris membentuk matriks yang nilainya selalu bertambah satu.
4. **`print()`** — Setelah perulangan dalam selesai, pindah ke baris berikutnya sehingga terbentuk pola matriks.

Dengan input `5` dihasilkan 5 baris dan 6 kolom (karena `range(batas_3015 + 1)` menghasilkan 6 nilai) dengan bilangan yang terus naik: baris pertama `1 2 3 4 5 6`, baris kedua `2 3 4 5 6 7`, dan seterusnya.


### 2.2.7 Program nested_for4_2611533015 (Nested For Pola Jam Pasir Berbingkai)


````python
# Buat file dengan nama nested_for4_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

tinggi_3015 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_3015 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_3015 = tinggi_3015
    c_3015 = a_3015
    lebar_3015 = (2 * tinggi_3015) - 2

    for i in range(1, tinggi_3015 + 1):
        b_3015 = c_3015 + 1

        for j in range(1, lebar_3015 + 1):

            # Baris atas dan bawah
            if i == 1 or i == tinggi_3015:
                if j == 1 or j == lebar_3015:
                    print("#", end="")
                else:
                    print("=", end="")
            # Baris isi
            else:
                if j == 1 or j == lebar_3015:
                    print("|", end="")
                else:
                    if j == c_3015:
                        print("<", end="")
                    elif j == b_3015:
                        print(">", end="")
                    elif j == (lebar_3015 - c_3015):
                        print("<", end="")
                    elif j == (lebar_3015 - c_3015 + 1):
                        print(">", end="")
                    elif j > b_3015 and j < (lebar_3015 - c_3015):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        # Logika asli Java
        a_3015 -= 2

        if a_3015 <= 0:
            c_3015 = (-a_3015) + 2
        else:
            c_3015 = a_3015
````

**Penjelasan Program nested_for4_2611533015:**

Program ini mendemonstrasikan **perulangan bersarang yang dikombinasikan dengan percabangan if-elif-else** untuk membentuk pola jam pasir berbentuk berlian di dalam bingkai.

1. **`tinggi_3015 = int(input(...))`** — Menerima tinggi pola; program hanya memproses bilangan genap karena pola jam pasir simetris hanya terbentuk sempurna dengan tinggi genap.
2. **`lebar_3015 = (2 * tinggi_3015) - 2`** — Lebar pola ditentukan dari tinggi. Untuk tinggi 10, lebar = 18.
3. **`a_3015` dan `c_3015`** — Menyimpan posisi pusat karakter. `c_3015` menentukan kolom tempat karakter `<` dan `>` dicetak, sedangkan `b_3015 = c_3015 + 1` menentukan kolom sebelahnya. Pada setiap baris nilai `a_3015` berkurang 2; ketika `a_3015 <= 0`, posisi pusat berbalik arah (`c_3015 = (-a_3015) + 2`) sehingga pola melebar kembali — inilah inti pembentukan bentuk jam pasir.
4. **`if j == 1 or j == lebar_3015:`** — Kolom pertama dan terakhir pada baris isi menjadi garis vertikal `|`; pada baris pertama dan terakhir menjadi sudut bingkai `#`.
5. **Percabangan `j == c_3015`, `j == b_3015`, `j == (lebar_3015 - c_3015)`** — Mencetak karakter `<` dan `>`, sedangkan kondisi `j > b_3015 and j < (lebar_3015 - c_3015)` mencetak titik `.` sebagai isi dan sisanya dicetak spasi.

Program ini menegaskan bahwa kombinasi nested for dan percabangan mampu menghasilkan pola geometri yang kompleks dan menjadi jembatan menuju studi kasus pada tugas praktikum.

### a. Kode Program tugas5_no1_2611533015 (Pola Segitiga Bintang)


````python
# Tugas output perulangan for
# Buat program untuk perulangan for dalam Python (Pola Segitiga Bintang)
# Program ini menggunakan fungsi input()

tinggi = int(input("Masukkan tinggi segitiga: "))

for i in range(1, tinggi + 1):
    # Cetak spasi di sebelah kiri agar membentuk segitiga simetris
    for j in range(tinggi - i):
        print(" ", end="")

    # Cetak bintang beserta spasi setelahnya
    for k in range(i):
        print("* ", end="")

    # Pindah ke baris baru
    print()
````

**Penjelasan Program tugas5_no1_2611533015 (Pola Segitiga Bintang):**

Program ini memenuhi Tugas 5 nomor 1, yaitu membuat program yang mengeluarkan output berupa segitiga menggunakan **tiga perulangan for** yang saling bekerja sama.

1. **`tinggi = int(input("Masukkan tinggi segitiga: "))`** — Menerima tinggi segitiga dari pengguna lalu mengonversinya menjadi integer.
2. **`for i in range(1, tinggi + 1):`** — Perulangan luar yang mengatur baris; dengan input `5`, baris berjalan dari `1` sampai `5`.
3. **`for j in range(tinggi - i): print(" ", end="")`** — Nested for pertama mencetak spasi di sebelah kiri sebanyak `tinggi - i`. Semakin besar `i`, semakin sedikit spasi, sehingga bintang bergeser ke kiri dan membentuk sisi miring segitiga.
4. **`for k in range(i): print("* ", end="")`** — Nested for kedua mencetak pasangan `"* "` sebanyak `i` kali, sehingga setiap baris memiliki jumlah bintang yang terus bertambah.
5. **`print()`** — Setelah satu baris selesai, pindah ke baris berikutnya.

Hasilnya adalah segitiga simetris dengan puncak satu bintang di atas dan alas lima bintang di bawah. Program ini menunjukkan bagaimana kombinasi nested for untuk spasi dan bintang menghasilkan pola geometri dua dimensi yang rapi.


### b. Kode Program tugas5_no2_2611533015 (Pola Jam Pasir Kristal Palindromik Berbingkai)


````python
print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")
n_3015 = int(input("Masukkan ukuran skala jam pasir (N): "))

# ---------- Bingkai atas ----------
print("#", end="")
for kolom_3015 in range(4 * n_3015 + 5):
    print("=", end="")
print("#", end="")
print()

# ---------- Fase 1: Jam pasir atas (baris N turun s.d. 1) ----------
for baris_3015 in range(n_3015, 0, -1):
    print("| ", end="")
    for spasi_3015 in range(2 * (n_3015 - baris_3015)):
        print(" ", end="")
    for angka_3015 in range(baris_3015, 0, -1):
        print(angka_3015, end=" ")
    print("<*>", end="")
    for angka_3015 in range(1, baris_3015 + 1):
        print(" ", end="")
        print(angka_3015, end="")
    for spasi_3015 in range(2 * (n_3015 - baris_3015)):
        print(" ", end="")
    print(" |", end="")
    print()

# ---------- Fase 2: Poros titik pusat ----------
print("|", end="")
for spasi_3015 in range(2 * n_3015 + 1):
    print(" ", end="")
print("<*>", end="")
for spasi_3015 in range(2 * n_3015 + 1):
    print(" ", end="")
print("|", end="")
print()

# ---------- Fase 3: Jam pasir bawah (baris 1 naik s.d. N) ----------
for baris_3015 in range(1, n_3015 + 1):
    print("| ", end="")
    for spasi_3015 in range(2 * (n_3015 - baris_3015)):
        print(" ", end="")
    for angka_3015 in range(baris_3015, 0, -1):
        print(angka_3015, end=" ")
    print("<*>", end="")
    for angka_3015 in range(1, baris_3015 + 1):
        print(" ", end="")
        print(angka_3015, end="")
    for spasi_3015 in range(2 * (n_3015 - baris_3015)):
        print(" ", end="")
    print(" |", end="")
    print()

# ---------- Bingkai bawah ----------
print("#", end="")
for kolom_3015 in range(4 * n_3015 + 5):
    print("=", end="")
print("#", end="")
print()
````

**Penjelasan Program tugas5_no2_2611533015 (Pola Jam Pasir Kristal Palindromik Berbingkai):**

Program ini adalah jawaban Studi Kasus Tugas 5 bagian B. Program mencetak pola **jam pasir kristal palindromik** yang dibingkai pagar `#` dan `=`, berisi deret angka menurun-menaik (palindromik) pada tiap baris dengan poros kristal `<*>` di tengahnya.

1. **`n_3015 = int(input(...))`** — Menerima skala `N`; pada contoh dijalankan dengan N = 4, sehingga angka tertinggi pada puncak dan alas adalah 4.
2. **Bingkai atas dan bawah** — `print("#", end="")` diikuti `for kolom_3015 in range(4 * n_3015 + 5)` yang mencetak `=` dan ditutup `#`, menghasilkan baris `#====...====#` dengan lebar bergantung N.
3. **Fase 1 (bagian atas)** — `for baris_3015 in range(n_3015, 0, -1)` mencetak baris dari N turun ke 1. Spasi kiri dicetak sebanyak `2 * (n_3015 - baris_3015)`; lalu deret angka menurun ditulis `for angka_3015 in range(baris_3015, 0, -1)`, kemudian poros `"<*>"`, dan deret angka menaik `for angka_3015 in range(1, baris_3015 + 1)`. Inilah pembentukan palindromik: `4 3 2 1 <*> 1 2 3 4`.
4. **Fase 2 (poros pusat)** — Baris tengah hanya berisi `"<*>"` yang diapit `2 * n_3015 + 1` spasi di kiri dan kanan, ditandai `|` pada kedua ujung.
5. **Fase 3 (bagian bawah)** — `for baris_3015 in range(1, n_3015 + 1)` mencetak baris dari 1 naik ke N, membalik urutan bagian atas sehingga bangun melebar kembali ke bawah, membentuk simetri jam pasir.

Program ini memadukan nested for untuk spasi, deret angka menurun-menaik, serta percabangan logika (pewarnaan bingkai) yang semuanya dibangun di atas konsep perulangan for.


### c. Jawaban Pertanyaan Analisis

**1. Bagaimana perulangan for digunakan untuk mencetak bingkai berpagar pada program?**

Bingkai atas dan bawah dicetak dengan `print("#", end="")` diikuti perulangan `for kolom_3015 in range(4 * n_3015 + 5): print("=", end="")` lalu ditutup `print("#", end="")`. Parameter `end=""` menjaga seluruh karakter tetap pada baris yang sama sehingga terbentuk satu baris penuh `#====...====#`. Lebar bingkai ditentukan oleh nilai `4 * n_3015 + 5`, sehingga semakin besar `n_3015`, semakin panjang bingkai `=` yang dihasilkan.

**2. Jelaskan bagaimana deret angka palindromik dibentuk pada setiap baris!**

Pada setiap baris, angka dicetak dua kali dalam pola cermin: pertama menurun dari `baris_3015` sampai `1` melalui `for angka_3015 in range(baris_3015, 0, -1)`, kemudian poros kristal `"<*>"`, lalu angka menaik dari `1` sampai `baris_3015` melalui `for angka_3015 in range(1, baris_3015 + 1)`. Karena bagian kiri dan kanan poros saling bercermin, deret yang terbentuk bersifat palindromik, misalnya `4 3 2 1 <*> 1 2 3 4`.

**3. Apa fungsi variabel `n_3015` dan bagaimana ia memengaruhi lebar bingkai serta poros pusat?**

`n_3015` adalah skala jam pasir sekaligus nilai tertinggi deret angka pada puncak dan alas. Nilai ini menentukan jumlah `=` pada bingkai (`4 * n_3015 + 5`), jumlah spasi di kiri dan kanan poros pusat (`2 * n_3015 + 1`), banyaknya spasi pergeseran baris (`2 * (n_3015 - baris_3015)`), serta banyaknya baris pada bagian atas dan bawah jam pasir. Semakin besar N, semakin lebar dan tinggi pola yang dihasilkan.

**4. Mengapa bagian bawah jam pasir dapat dicetak hanya dengan mengubah arah range pada perulangan luar?**

Bagian atas menggunakan `for baris_3015 in range(n_3015, 0, -1)` (menghitung mundur dari N ke 1), sedangkan bagian bawah menggunakan `for baris_3015 in range(1, n_3015 + 1)` (menghitung maju dari 1 ke N). Logika pembentukan setiap baris identik; perbedaan arah perulangan luar inilah yang membuat jam pasir mencekung lalu melebar kembali, sehingga terbentuk simetri terhadap poros pusat.

**5. Apa peran parameter end="" pada seluruh perintah print() pembentuk baris?**

Parameter `end=""` menggantikan default pemindahan baris (`\n`) pada `print()`, sehingga seluruh karakter seperti pagar, spasi, angka, dan poros dapat dirangkai menjadi satu baris pola yang utuh. Pemindahan baris dilakukan secara eksplisit hanya melalui `print()` tanpa argumen di akhir setiap baris, sehingga setiap baris pola tercetak pada baris tersendiri.

**6. Identifikasi baris paling atas dan baris poros pusat pada contoh output N = 4!**

Pada output N = 4, baris paling atas adalah `| 4 3 2 1 <*> 1 2 3 4 |` yang memuat deret palindromik terpanjang; baris poros pusat adalah `|         <*>         |` yang hanya berisi poros kristal `<*>` yang diapit 9 spasi di kiri dan 9 spasi di kanan (`2 * N + 1 = 9`). Keduanya menunjukkan bagaimana struktur jam pasir menyatu antara deret angka dinamis dan poros kristal di pusat bangun.
