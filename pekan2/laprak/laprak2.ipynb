## 2.1 Dasar Teori

Praktikum Tipe Data dan Variabel membahas berbagai jenis tipe data dalam Python serta cara mendeklarasikan dan menggunakan variabel.

### A. Tipe Data Numerik
Terdiri dari tiga jenis: **int** (bilangan bulat), **float** (bilangan desimal), dan **complex** (bilangan kompleks a + bj).

### B. Tipe Data String
String dapat ditulis dengan petik satu (`'...'`), petik dua (`"..."`), atau tiga petik (`"""..."""`) untuk multiline.

### C. Tipe Data Boolean
Hanya memiliki dua nilai: **True** dan **False**. Dihasilkan dari operator perbandingan dan digunakan bersama operator logika (`and`, `or`, `not`).

### D. Fungsi Input dan Konversi Tipe Data
Fungsi `input()` mengembalikan nilai string. Konversi dengan `int()` atau `float()` diperlukan untuk operasi matematika.

### E. Konstanta (Final)
Konstanta didefinisikan dengan `Final` dari modul `typing`, nilainya tetap dan tidak dapat diubah.

## 2.2 Kode Program

### 2.2.1 Program Numeric_2611533015


```python
#create a variable with integer value
a_3015 = 100
print("The type of variable having value",a_3015,"is",type(a_3015))

#create variable with float value
b_3015 = 10.2345
print("The type of variable having value",b_3015,"is",type(b_3015))

#create variable with complex value
c_3015 = 100 + 3j
print("The type of variable having value",c_3015,"is",type(c_3015))
```

**Penjelasan Program Numeric_2611533015:**

Program ini mendemonstrasikan tiga jenis tipe data numerik dalam Python, yaitu integer, float, dan complex.

1. **`a_3015 = 100`** - Variabel integer (bilangan bulat).
2. **`type(a_3015)`** - Fungsi `type()` mengecek tipe data, hasilnya `<class 'int'>`.
3. **`b_3015 = 10.2345`** - Variabel float (bilangan pecahan).
4. **`c_3015 = 100 + 3j`** - Variabel complex (bilangan kompleks).
5. Fungsi `print()` menampilkan teks dan nilai variabel ke layar.

Program ini menunjukkan bahwa Python menentukan tipe data secara otomatis. Fungsi `type()` berguna untuk memeriksa tipe data variabel.

### 2.2.2 Program String_2611533015


```python
# Buat file dengan nama string String_2611533015.py
# nama variabel ditambah 4 digit terakhir contoh: nama_3015
# ganti programnya dan data masing2
nama_3015 = "Zikri"
jenis_kelamin_3015 = 'Laki-Laki'
alamat_3015 = """
    ASR TNI-AD GANTING,
    Ganting, Padang,
    Sumatera Barat, Indonesia
"""

agama_3015 = "Islam"
print("Nama saya adalah: ", nama_3015)
print("Jenis kelamin saya adalah: ", jenis_kelamin_3015)
print("Alamat saya adalah: ", alamat_3015)
print("Agama saya adalah: ", agama_3015)
```

**Penjelasan Program String_2611533015:**

Program ini mendemonstrasikan tipe data **string** (teks) dan berbagai cara penulisannya.

1. **`nama_3015 = "Zikri"`** — Variabel bertipe string yang ditulis menggunakan tanda petik dua ("..."), berisi nama pemilik program.
2. **`jenis_kelamin_3015 = 'Laki-Laki'`** — Variabel string yang ditulis menggunakan tanda petik satu ('...'). Python memperlakukan petik satu dan petik dua secara setara untuk string biasa,    sehingga keduanya bisa dipakai sesuai preferensi penulis kode.
3. **`alamat_3015 = """..."""`** — Variabel string yang ditulis menggunakan tiga tanda petik ("""..."""), yang disebut triple-quoted string. Format ini digunakan khusus untuk teks yang terdiri dari beberapa baris (multiline), sehingga baris baru (enter) di dalam teks ikut tersimpan sebagai bagian dari string, tanpa perlu karakter escape seperti \n.
4. **`agama_3015 = "Islam"`**` — Variabel string sederhana yang ditulis dengan petik dua, berisi data agama.
5. **`Fungsi print()`** digunakan sebanyak empat kali untuk menampilkan masing-masing label teks yang diikuti oleh nilai variabelnya. Ketika print() diberi dua argumen yang dipisahkan koma (misalnya print("Nama saya adalah: ", nama_3015)), Python otomatis menyisipkan satu spasi di antara kedua argumen tersebut saat ditampilkan ke layar.

Program ini menunjukkan fleksibilitas penulisan string dalam Python.

### 2.2.3 Program Boolean_2611533015


```python
# Buar file dengan nama Boolean_2611533015
# Nama variabel ditambah 4 digit nim terakhir contoh: nilai_1234
# Deklarasi variabel dengan tipe data Boolean
is_lulus_3015 = True
is_cumlaude_3015 = True

# Menggunakan Boolean
nilai_3015 = 85
batas_lulus_3015 = 75

# Menantukan nilai Boolean dari kondisi
status_kelulusan_3015 = nilai_3015 >= batas_lulus_3015 # Hasilnya akan True

print("=== Check Kelulusan ===")
print("Nilai: ",nilai_3015)
print("Apakah lulus?: ",status_kelulusan_3015)
if is_lulus_3015 and is_cumlaude_3015:
    print("Selamat, Anda lulus dengan predikat Cum Laude!")
```

**Penjelasan Program Boolean_2611533015:**

Program ini mendemonstrasikan tipe data **boolean** (True/False) dan penerapannya dalam logika dan percabangan.

1. **`is_lulus_3015 = True dan is_cumlaude_3015 = True`** — Kedua variabel ini dideklarasikan langsung dengan tipe data boolean, yaitu tipe data yang hanya memiliki dua kemungkinan nilai: True (benar) atau False (salah). Di Python, boolean sebenarnya adalah bentuk khusus dari tipe integer, di mana True bernilai 1 dan False bernilai 0.
2. **`nilai_3015 = 85 dan batas_lulus_3015 = 75`** — Merupakan variabel bertipe integer yang digunakan sebagai data pembanding, yaitu nilai yang diperoleh dan batas minimum kelulusan.
3. **`status_kelulusan_3015 = nilai_3015 >= batas_lulus_3015`** — Baris ini menggunakan operator perbandingan >= (lebih besar atau sama dengan) untuk membandingkan nilai_3015 (85) dengan batas_lulus_3015 (75). Karena 85 lebih besar dari 75, ekspresi ini menghasilkan nilai boolean True, yang kemudian disimpan ke dalam variabel status_kelulusan_3015. Ini menunjukkan bahwa hasil dari sebuah operator perbandingan selalu berupa boolean.
4. **`print(...)`** pada baris-baris berikutnya menampilkan judul program serta nilai dari nilai_3015 dan status_kelulusan_3015 ke layar, sehingga pengguna bisa melihat langsung hasil perbandingan yang telah dilakukan.
5. **`if is_lulus_3015 and is_cumlaude_3015:`** — Baris ini merupakan struktur percabangan if yang kondisinya menggunakan operator logika and. Operator and akan menghasilkan True hanya jika kedua operand-nya bernilai True. Karena is_lulus_3015 dan is_cumlaude_3015 sama-sama True, maka kondisi keseluruhan bernilai True, sehingga blok kode di dalam if akan dieksekusi.
6. Ketika kondisi pada if terpenuhi, program akan mencetak pesan "Selamat, Anda lulus dengan predikat Cum Laude!" ke layar sebagai output akhir dari program ini.

Boolean berguna dalam pengambilan keputusan dengan `if`. Operator perbandingan dan logika menghasilkan boolean.

### 2.2.4 Program Input_2611533015


```python
# Buat file dengan nama Input_261153015.py
# Buat Program untuk menghitung luas persegi panjang
# Nama variabel ditambah 4 digit mnimm terakhir contoh: nama_3015
# Program ini menggunakan fungsi input() untuk meminta pengguna
# memasukan nilai panjang dan lebar persegi panjang
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
panjang_3015 = input("Masukkan nilai panjang: ")
lebar_3015 = input("Masukkan nilai lebar: ")
luas_3015 = int(panjang_3015) * int(lebar_3015)
print("Luas = ",luas_3015)
```

**Penjelasan Program Input_2611533015:**

Program ini mendemonstrasikan fungsi **`input()`** untuk menerima masukan pengguna dan **konversi tipe data**.

1. **`panjang_3015 = input("Masukkan nilai panjang: ")`** — Fungsi input() menampilkan teks "Masukkan nilai panjang: " sebagai prompt, lalu program berhenti sejenak menunggu pengguna mengetikkan nilai dan menekan Enter. Nilai yang dimasukkan pengguna akan disimpan ke dalam variabel panjang_3015.
2. **`lebar_3015 = input("Masukkan nilai lebar: ")`** — Sama seperti baris sebelumnya, tetapi untuk meminta nilai lebar dari pengguna.
3. **`Penting: fungsi input()`** selalu mengembalikan nilai bertipe string, meskipun pengguna mengetik angka. Jadi jika pengguna mengetik 5, nilai yang tersimpan di panjang_3015 bukanlah angka 5, melainkan teks "5".
4. **`luas_3015 = int(panjang_3015) * int(lebar_3015)`** — Karena panjang_3015 dan lebar_3015 masih bertipe string, keduanya tidak bisa langsung dikalikan secara matematis. Oleh karena itu, fungsi int() digunakan untuk mengonversi (type casting) nilai string tersebut menjadi tipe integer, barulah operator * bisa melakukan perkalian matematis yang sebenarnya, dan hasilnya disimpan ke variabel luas_3015.
5. **`print("Luas = ", luas_3015)`** — Menampilkan hasil perhitungan luas ke layar.




**Catatan:** `input()` selalu mengembalikan string. Konversi dengan `int()`/`float()` diperlukan untuk operasi matematika.

### 2.2.5 Program Konstanta_2611533015


```python
# Buat file dengan nama Konstanta_2611533015
# Program ini menggunakan konstanta untuk menghitun luas lingkaran
# nama variabel ditambah 4 digit nim terakhir contoh: jari_3015

from typing import Final
PI: Final = 3.14
print("Pi: %f" % (PI))
jari_3015 = float(input("Masukkan nilai jari-jari: "))
luas_3015 = PI * jari_3015
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_3015,luas_3015))
```

**Penjelasan Program Konstanta_2611533015:**

Program ini mendemonstrasikan **konstanta** dengan `Final` dari modul `typing` untuk menghitung luas lingkaran.

1. **`from typing import Final`** — Baris ini mengimpor Final dari modul typing. Final digunakan sebagai penanda (type hint) bahwa sebuah variabel dimaksudkan sebagai konstanta, yaitu nilai yang seharusnya tidak diubah lagi setelah didefinisikan.
2. **`PI: Final = 3.14`** — Mendeklarasikan variabel PI dengan anotasi tipe Final dan memberinya nilai 3.14. Perlu dicatat bahwa Final di Python hanya bersifat penanda bagi programmer dan tools seperti type checker (misalnya mypy) — Python sendiri secara teknis tidak benar-benar mencegah nilai PI diubah di runtime, tapi konvensi penamaan huruf kapital (PI) dan penanda Final memberi sinyal jelas bahwa variabel ini tidak boleh diubah.
3. **`print("Pi: %f" % (PI))`** — Menggunakan operator format string %, salah satu cara lama (namun masih umum dipakai) untuk menyisipkan nilai variabel ke dalam teks. Kode format %f khusus digunakan untuk menampilkan nilai bertipe float.
4. **`jari_3015 = float(input("Masukkan nilai jari-jari: "))`** — Baris ini meminta input dari pengguna menggunakan input() (yang hasilnya berupa string), lalu langsung membungkusnya dengan float() agar nilainya otomatis dikonversi menjadi tipe data float, sehingga bisa langsung digunakan dalam perhitungan matematis tanpa perlu variabel perantara.
5. **`luas_3015 = PI * jari_3015`** — Menghitung luas lingkaran menggunakan rumus sederhana PI * jari-jari (catatan: ini merupakan penyederhanaan dari rumus luas lingkaran yang sebenarnya π × r², karena pada kode ini jari-jari tidak dipangkatkan dua).
6. **`print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_3015, luas_3015))`** — Menampilkan hasil akhir menggunakan format %.2f, yaitu kode format untuk menampilkan bilangan float dengan 2 angka di belakang koma, sehingga tampilan angka lebih rapi dan mudah dibaca.

Konstanta berguna untuk menyimpan nilai tetap, membuat program lebih rapi.
